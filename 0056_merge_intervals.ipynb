# Problem 

# 56: Merge Intervals

# Description

Given an array of intervals where `intervals[i] = [starti, endi]`, merge all overlapping intervals and return an array of the non-overlapping intervals that cover all the intervals in the input.

# Examples

## Example 1

**Input:**  
`intervals = [[1,3],[2,6],[8,10],[15,18]]`  
**Output:**  
`[[1,6],[8,10],[15,18]]`  
**Explanation:**  
Since intervals `[1,3]` and `[2,6]` overlap, merge them into `[1,6]`.

## Example 2

**Input:**  
`intervals = [[1,4],[4,5]]`  
**Output:**  
`[[1,5]]`  
**Explanation:**  
Intervals `[1,4]` and `[4,5]` are considered overlapping.

# Constraints

- $ 1 \leq \text{intervals.length} \leq 10^4$ 
- $ \text{intervals}[i].\text{length} = 2$ 
- $ 0 \leq \text{starti} \leq \text{endi} \leq 10^4$ 

```python
# intervals = [[1,3], [8,10], [9, 11], [14, 17], [15,18], [2,6]]
# # intervals = [[1,4],[0,1]]
intervals = [[2,3],[4,5],[6,7],[8,9],[1,10]]
```

# Using Start Values for Sorting

```python

intervals.sort(key = lambda x: x[0])

new_intervals = []
new_intervals.append(intervals[0].copy())

for current_interval in intervals:
    # if current interval doesn't overlap with the previuosly handled merged interval
    # append it
    if current_interval[0] > new_intervals[-1][1]:
        print("no overlap")
        print(f"Added {current_interval}")
        new_intervals.append(current_interval)
    else:
        new_end     = max(new_intervals[-1][1], current_interval[1])
        new_start   = new_intervals[-1][0]
        new_intervals[-1][0] = new_start
        new_intervals[-1][1] = new_end
        print("overlap")
        print(f"Updated interval {new_intervals[-1]} with {new_start} and {new_end} values")

    
    # if current_interval[0] <= first_end:
    #     overlap = 1
    #     print("overlap")
    #     first_end = current_interval[1]
    #     new_interval = (interval_start, first_end)
    #     new_intervals.pop()
    #     new_intervals.append(new_interval)
    # else:
    #     print("no overlap")
    #     new_intervals.append(current_interval)
    #     first_end = current_interval[1]
```
