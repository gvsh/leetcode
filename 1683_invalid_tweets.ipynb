# Problem 1683: Invalid Tweets

## Description
Table: **Tweets**

| Column Name | Type    |
|-------------|---------|
| tweet_id    | int     |
| content     | varchar |

- `tweet_id` is the primary key for this table.
- The `content` column consists of characters from an American keyboard and contains no special characters.
- This table contains all tweets in a social media app.

The task is to find the IDs of tweets that are invalid. A tweet is considered invalid if the number of characters in the `content` column exceeds 15.

Return the result in any order.

## Examples

### Example 1

**Input:**  
**Tweets** table:

| tweet_id | content                           |
|----------|-----------------------------------|
| 1        | Let us Code                       |
| 2        | More than fifteen chars are here! |

**Output:**  

| tweet_id |
|----------|
| 2        |

**Explanation:**  
- Tweet 1 has a length of 11 characters and is valid.
- Tweet 2 has a length of 33 characters and is invalid.

## Constraints
- `tweet_id` is guaranteed to be unique.
- The length of the `content` column can vary but consists only of American keyboard characters.

## Follow Up
What would be the optimal query to identify invalid tweets if the table had millions of entries?


# Implementation

```python
import random
```

```python
import string
```

```python
string.ascii_letters
```

```python
string.digits
```

```python
def get_keyboard_chars():
    keyboard_chars = []
    for i in range(32, 127):  
        keyboard_chars.append(chr(i))
    keyboard_chars.append(" ")
    return keyboard_chars

keyboard_chars = get_keyboard_chars()
print(keyboard_chars)
```

```python
tweet_id_list = list(range(50))
```

```python
def get_n_tweets(n=50):
    return ["".join([random.choice(keyboard_chars) for _ in range(random.randint(1, 45))]) for _ in range(n)]
```

```python
tweet_content_list = get_n_tweets()
```

```python
tweet_content_list
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(list(zip(tweet_id_list, tweet_content_list)), columns=['tweet_id', 'tweet_content'])
```

```python
df.head()
```

```python
df.loc[df.loc[:, 'tweet_content'].str.len() <= 15, ['tweet_id']]
```

```python
import pandas as pd

def invalid_tweets(tweets: pd.DataFrame) -> pd.DataFrame:

    return tweets.loc[tweets.loc[:, 'content'].str.len() > 15, ['tweet_id']]
    
```
