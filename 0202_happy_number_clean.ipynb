# 202. Happy Number

---

### Problem

Write an algorithm to determine if a number `n` is happy.

A **happy number** is a number defined by the following process:

1. Starting with any positive integer, replace the number by the sum of the squares of its digits.
2. Repeat the process until the number equals 1 (where it will stay), or it loops endlessly in a cycle which does not include 1.

Those numbers for which this process ends in 1 are happy.

Return `true` if `n` is a happy number, and `false` if not.

---

### Example 1

**Input:**  
`n = 19`  

**Output:**  
`true`  

**Explanation:**  
- $1^2 + 9^2 = 82$  
- $8^2 + 2^2 = 68$  
- $6^2 + 8^2 = 100$  
- $1^2 + 0^2 + 0^2 = 1$

---

### Example 2

**Input:**  
`n = 2`  

**Output:**  
`false`

---

### Constraints

- $1 \leq n \leq 2^{31} - 1$

# Steps

1. For the input number, get all digits
2. Sum of squares of each of digits
3. If this sum of squares has already been seen, then we have entered a cycle, quit with failure
4. If this sum of squares is 1, quit with success
5. This becomes the new number, go to step 1

```python
class Solution:
    def isHappy(self, n: int) -> bool:
        
        def get_new_digits(input_number):
            
            digit_array = []
            
            while input_number > 0:
                input_number, digit = divmod(input_number, 10)
                digit_array.append(digit)
            
            return digit_array  
        
        def get_sum_of_squares(digit_array):
        
            return sum([x * x for x in digit_array])

        
        seen_sum_of_squares = set()
        random_number = n
        
        while True:
                
            digit_array    = get_new_digits(random_number)
            sum_of_squares = get_sum_of_squares(digit_array)
            
            if sum_of_squares == 1:
                return True
            elif sum_of_squares in seen_sum_of_squares:
                return False
            else:
                seen_sum_of_squares.add(sum_of_squares)
                random_number = sum_of_squares
        
```

```python
# random_number = 2
random_number = random.randint(1, 1000)
print(f"{random_number=}")
```

```python

seen_sum_of_squares = set()
return_value = -1
counter = 0

while counter < 200:
        
    digit_array    = get_new_digits(random_number)
    print(f"{digit_array=}")
    sum_of_squares = get_sum_of_squares(digit_array)
    print(f"{digit_array=}")
    
    if sum_of_squares == 1:
        print("Happy Number")
        return_value = 1
        break
    elif sum_of_squares in seen_sum_of_squares:
        print("Cycle detected")
        return_value = 0
        break
    else:
        seen_sum_of_squares.add(sum_of_squares)
        random_number = sum_of_squares
        print(f"New {random_number=}")
        print("New value seen")
        counter += 1
        print(f"New {counter=}")

print(f"{return_value=}")
```
