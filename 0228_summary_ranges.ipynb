# Problem

# 228. Summary Ranges

**Easy**

---

# Description

You are given a sorted unique integer array `nums`.

A range `[a,b]` is the set of all integers from `a` to `b` (inclusive).

Return the smallest sorted list of ranges that cover all the numbers in the array exactly. That is, each element of `nums` is covered by exactly one of the ranges, and there is no integer `x` such that `x` is in one of the ranges but not in `nums`.

Each range `[a,b]` in the list should be output as:

- `"a->b"` if `a != b`  
- `"a"` if `a == b`

---

# Examples

## Example 1

**Input:**  
`nums = [0,1,2,4,5,7]`  

**Output:**  
`["0->2","4->5","7"]`  

**Explanation:**  
The ranges are:  
- `[0,2]` --> `"0->2"`  
- `[4,5]` --> `"4->5"`  
- `[7,7]` --> `"7"`

---

## Example 2

**Input:**  
`nums = [0,2,3,4,6,8,9]`  

**Output:**  
`["0","2->4","6","8->9"]`  

**Explanation:**  
The ranges are:  
- `[0,0]` --> `"0"`  
- `[2,4]` --> `"2->4"`  
- `[6,6]` --> `"6"`  
- `[8,9]` --> `"8->9"`

---

# Constraints

- $ 0 \leq nums.length \leq 20$   
- $ -2^{31} \leq nums[i] \leq 2^{31} - 1$   
- All the values of `nums` are unique.  
- `nums` is sorted in ascending order.

# Solution

```python
nums = [0, 1, 2, 4, 5, 7]
```

1. Let's create a new array with tuples representing the start and end values of the range.
2. At initialization, these start and end values for a particular entry will be the same.
3. We then start merging these.

```python
i = 0

previous_interval = [nums[0]]
all_intervals     = [previous_interval]

while i < (len(nums) - 1):
    if (nums[i + 1] - nums[i]) == 1:
        new_interval = previous_interval.append(nums[i + 1])
    else:
        new_interval = [nums[i + 1]]
        all_intervals.append(new_interval)
        previous_interval = new_interval
    i = i + 1
```

```python
returned_intervals = []

for interval in all_intervals:
    if len(interval) > 1:
        current_string = str(interval[0]) + "->" + str(interval[-1])
    else:
        current_string = str(interval[0])
    returned_intervals.append(current_string)
        
```

```python
from typing import List
```

```python
class Solution:
    def summaryRanges(self, nums: List[int]) -> List[str]:
        
        i = 0
        
        if len(nums) > 0:
            previous_interval = [nums[0]]
            all_intervals     = [previous_interval]

            while i < (len(nums) - 1):
                if (nums[i + 1] - nums[i]) == 1:
                    new_interval = previous_interval.append(nums[i + 1])
                else:
                    new_interval = [nums[i + 1]]
                    all_intervals.append(new_interval)
                    previous_interval = new_interval
                i = i + 1

            returned_intervals = []

            for interval in all_intervals:
                if len(interval) > 1:
                    current_string = str(interval[0]) + "->" + str(interval[-1])
                else:
                    current_string = str(interval[0])
                returned_intervals.append(current_string)

            return returned_intervals
        else:
            return []                            
```

```python
solution = Solution()
solution.summaryRanges(nums)
```
