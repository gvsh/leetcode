# Problem 102: Binary Tree Level Order Traversal

# Description
Given the root of a binary tree, return the level order traversal of its nodes' values. This traversal collects values level by level from left to right.

# Examples

## Example 1

**Input:**  
`root = [3,9,20,null,null,15,7]`  
**Output:**  
`[[3],[9,20],[15,7]]`  

## Example 2
**Input:**  
`root = [1]`  
**Output:**  
`[[1]]`  

## Example 3
**Input:**  
`root = []`  
**Output:**  
`[]`  

# Constraints

- The number of nodes in the tree is in the range [0, 2000].
- -1000 <= Node.val <= 1000
