# Problem

# 383 Ransom Note

Given two strings `ransomNote` and `magazine`, return `true` if `ransomNote` can be constructed by using the letters from `magazine` and `false` otherwise.

Each letter in `magazine` can only be used once in `ransomNote`.

# Examples

**Example 1:**

```
ransomNote = "a", magazine = "b"
```
Output:
```
false
```

  


**Example 2:**

Input:
```
ransomNote = "aa", magazine = "ab"
```
Output:
```
false
```



**Example 3:**

Input:
```
ransomNote = "aa", magazine = "aab"
```
Output:
```
true
```

# Constraints


- $1$ `<= ransomNote.length, magazine.length <= `$10^5$
- `ransomNote` and `magazine` consist of lowercase English letters.


# Solution

````python
import random
from random import choice
from typing import List, Tuple, Union, Optional
````

````python
import string

print(string.ascii_letters)
````

````python
ascii_letters = string.ascii_letters
````

````python
def get_random_lower_case_word(max_length_of_word):

    s_length = random.randint(0, max_length_of_word)
    s        = "".join([choice(ascii_letters[:10]) for _ in range(s_length)])

    # return s.lower()
    return ("abcd" + s.lower())
````

````python
def get_random_word():

    s_length = random.randint(0, 15)
    s        = "".join([choice(ascii_letters) for _ in range(s_length)])

    return s
````

````python
single_random_word = get_random_lower_case_word(15)
print(single_random_word)
````

````python
def get_ransom_note_and_magazine_test_suite(number_of_test_cases):

    test_suite = []

    for x in range(number_of_test_cases):

        ransom_note = get_random_lower_case_word(10)
        magazine    = get_random_lower_case_word(40)

        test_tuple = ransom_note, magazine

        test_suite.append(test_tuple)

    return test_suite
````

````python
test_suite = get_ransom_note_and_magazine_test_suite(50)
````

````python
def can_construct(ransom_note, magazine):

    print(f"{ransom_note=}")
    print(f"{magazine=}")

    ransom_counter   = Counter(ransom_note)
    magazine_counter = Counter(magazine)

    print(f"{ransom_counter=}")
    print(f"{magazine_counter=}")

    for key, value in ransom_counter.items():

        print(key, value)
        if key in magazine_counter:
            if magazine_counter[key] >= value:
                pass
            else:
                print(f"Not enough occurences of {key} in magazine: Required {value}, found only {magazine_counter[key]} counts")
                print("*********************")
                print("*********************")
                print("*********************")
                return False
        else:
            print(f"{key} not found in magazine")
            print("*********************")
            print("*********************")
            print("*********************")
            return False

    print("@@@@@@@@@@@@@@@@@@@@@")
    print("@@@@@@@@@@@@@@@@@@@@@")
    print("@@@@@@@@@@@@@@@@@@@@@")

    return True
````
