```python
def remove_duplicates_from_sorted_array(nums):
    test_input           = nums
    n_dups               = 0
    i                    = 0
    ever_duplicate       = 0
    len_test_input       = len(test_input)
    while i <= (len_test_input - 2):
        # if i == 0:
        #     new_test_input[i] = test_input[i]
        # print(test_input[i])
        # print("**********")
        # print(f"Iteration number: {(i + 1)=}")
        # print("**********")
        increment = test_input[i + 1] - test_input[i]
        if increment == 0:
            n_dups += 1
            # print(f"{n_dups=}")
            # print(f"{i + 1} index should be deleted, value {test_input[i + 1]} same as previous index value {test_input[i]}:")
            ever_duplicate = 1
        else:
            # print(f"{n_uniques=}")
            if ever_duplicate == 1:
                # print(f"{i + 1} index should be moved by {n_dups} places, value {test_input[i + 1]} different than previous index value {test_input[i]}:")
                # new_test_input[i + 1 - n_dups] = test_input[i + 1]
                test_input[i + 1 - n_dups] = test_input[i + 1]
        i = i + 1
    
    return (len_test_input - n_dups), test_input
```
