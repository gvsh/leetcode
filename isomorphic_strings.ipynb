```python
import string
import random
```

```python
letters = [char for char in string.ascii_lowercase]
```

```python
letters.append(" ")
```

```python
print(letters)
```

```python
random_letter = None
new_word_chars = []

while True:
    random_letter = random.choice(letters)
    if random_letter != " ":
        new_word_chars.append(random_letter)
    else:
        break

new_word = "".join(new_word_chars)
```

```python
s_to_t_map    = {}
is_isomorphic = True

if len(s) != len(t):
    is_isomorphic = False
else:    
    for s_char, t_char in zip(s, t):
        print("*" * 80)
        print(f"{s_char=}, {t_char=}, {s_to_t_map}")
        print(f"First if: {s_char=}, {t_char=}, {s_to_t_map}")
        if s_char in s_to_t_map:
            print(f"Second if: {s_char=}, {t_char=}, {s_to_t_map}")
            if s_to_t_map[s_char] != t_char:
                print(f"Third if: {s_char=}, {t_char=}, {s_to_t_map}")
                is_isomorphic = False
        else:
            if t_char in s_to_t_map:
                if s_to_t_map[t_char] != s_char:
                    is_isomorphic = False
                    print(f"Else case break: {s_char=}, {t_char=}, {s_to_t_map}")
                    break
            else:
                s_to_t_map[s_char] = t_char
                print(f"Else case: {s_char=}, {t_char=}, {s_to_t_map}")
            
```
