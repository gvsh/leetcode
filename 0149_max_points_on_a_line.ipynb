# Problem 149. Max Points on a Line

# Description

Given an array of points where `points[i] = [x_i, y_i]` represents a point on the X-Y plane, return the maximum number of points that lie on the same straight line.

# Examples

## Example 1

**Input:** 

`points = [[1,1],[2,2],[3,3]]`


**Output:** 

`3`

## Example 2

**Input:** 

`points = [[1,1],[3,2],[5,3],[4,1],[2,3],[1,4]]`


**Output:** 

`4`

# Constraints

- $ 1 \leq \text{points.length} \leq 300 $
- $ \text{points[i].length} = 2 $
- $ -10^4 \leq x_i, y_i \leq 10^4 $
- All the points are unique.

```python
points = [[1,1],[3,2],[5,3],[4,1],[2,3],[1,4]]
points = [[0,0],[1,-1],[1,1]]
points = [[4,5],[4,-1],[4,0]]
```

```python
from collections import defaultdict

tan_of_angle = defaultdict(list)

for i in range(len(points)):
    for j in range(i + 1, len(points)):
        for k in range(j + 1, len(points)):
            point1   = points[i]
            point2   = points[j]
            point3   = points[k]
            [x1, y1] = point1
            [x2, y2] = point2
            [x3, y3] = point3
            
            print(f"{point1}, {point2} and {point3}")
            print(f"{((x1 - x2) * (y1 - y3))=}, {((x1 - x3) * (y1 - y2))=}")
            
            if (x1 - x2) * (y1 - y3) == (x1 - x3) * (y1 - y2):
                print(f"{point1}, {point2} and {point3} are in a line")
                angle = (y1 - y2) / (x1 - x2)
                if (x1, y1) not in tan_of_angle[angle]:
                    tan_of_angle[angle].append((x1, y1))
                if (x2, y2) not in tan_of_angle[angle]:
                    tan_of_angle[angle].append((x2, y2))
                if (x3, y3) not in tan_of_angle[angle]:
                    tan_of_angle[angle].append((x3, y3))
```

```python
def max_points(points):

    if len(points) < 2:
        return 1
    if len(points) == 2:
        return 2

    tan_of_angle     = defaultdict(list)
    vertical_lines   = defaultdict(list)
    horizontal_lines = defaultdict(list)

    for i in range(len(points)):
        for j in range(i + 1, len(points)):
            for k in range(j + 1, len(points)):

                
                point1   = points[i]
                point2   = points[j]
                point3   = points[k]
                [x1, y1] = point1
                [x2, y2] = point2
                [x3, y3] = point3

                
                if (x1 - x2) * (y1 - y3) == (x1 - x3) * (y1 - y2):
                    # print(f"{point1}, {point2} and {point3} are in a line")
                    if (x1 - x2) != 0:
                        
                        if (y1 - y3) == 0:
                            # print("zero case again")


                            
                            horizontal_line_y = y1
                            # print(f"{point1}, {point2} and {point3} are in a line")
                            if (x1, y1) not in horizontal_lines[horizontal_line_y]:
                                horizontal_lines[horizontal_line_y].append((x1, y1))
                            if (x2, y2) not in horizontal_lines[horizontal_line_y]:
                                horizontal_lines[horizontal_line_y].append((x2, y2))
                            if (x3, y3) not in horizontal_lines[horizontal_line_y]:
                                horizontal_lines[horizontal_line_y].append((x3, y3))  

                        
                        else:

                            
                            angle = (y1 - y2) / (x1 - x2)
                            if (x1, y1) not in tan_of_angle[angle]:
                                tan_of_angle[angle].append((x1, y1))
                            if (x2, y2) not in tan_of_angle[angle]:
                                tan_of_angle[angle].append((x2, y2))
                            if (x3, y3) not in tan_of_angle[angle]:
                                tan_of_angle[angle].append((x3, y3))       
                    else:
                        # print("zero case")
                        if (x1 - x3) == 0:
                            # print("zero case again")


                            
                            vertical_line_x = x1
                            # print(f"{point1}, {point2} and {point3} are in a line")
                            if (x1, y1) not in vertical_lines[vertical_line_x]:
                                vertical_lines[vertical_line_x].append((x1, y1))
                            if (x2, y2) not in vertical_lines[vertical_line_x]:
                                vertical_lines[vertical_line_x].append((x2, y2))
                            if (x3, y3) not in vertical_lines[vertical_line_x]:
                                vertical_lines[vertical_line_x].append((x3, y3))  

    # print(f"{tan_of_angle.items()=}")
    # print(f"{vertical_lines.items()=}")
    # print(f"{tan_of_angle.items()=}")
    # print(f"{horizontal_lines.items()=}")

    # for keys, values in vertical_lines.items():
    #     print(f"{keys=}, {len(values)=}")
    # for keys, values in tan_of_angle.items():
    #     print(f"{keys=}, {len(values)=}")
    # for keys, values in horizontal_lines.items():
    #     print(f"{keys=}, {len(values)=}")
    
    for keys, values in tan_of_angle.items():
        if keys == -1:
            print(f"{keys=}, {values=}")
            
    if tan_of_angle:
        max_of_slanted_lines = max([len(x[1]) for x in tan_of_angle.items()])
    else:
        max_of_slanted_lines = 0
        
    if vertical_lines:
        max_of_vertical_lines = max([len(x[1]) for x in vertical_lines.items()])
    else:
        max_of_vertical_lines = 0
        
    if horizontal_lines:
        max_of_horizontal_lines = max([len(x[1]) for x in horizontal_lines.items()])
    else:
        max_of_horizontal_lines = 0

    total_max = max(max_of_slanted_lines, max_of_vertical_lines, max_of_horizontal_lines)
    
    if tan_of_angle or vertical_lines or max_of_horizontal_lines:
        return total_max
    else:
        if len(points) > 2:
            return 2
```
