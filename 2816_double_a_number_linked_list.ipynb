# Problem 2816. Double a Number Represented as a Linked List

# Description
You are given the head of a non-empty linked list representing a non-negative integer without leading zeroes.

Return the head of the linked list after doubling it.

# Examples

## Example 1
**Input:**  
`head = [1,8,9]`  
**Output:**  
`[3,7,8]`  
**Explanation:**  
The figure above corresponds to the given linked list which represents the number `189`. Hence, the returned linked list represents the number `189 * 2 = 378`.

## Example 2
**Input:**  
`head = [9,9,9]`  
**Output:**  
`[1,9,9,8]`  
**Explanation:**  
The figure above corresponds to the given linked list which represents the number `999`. Hence, the returned linked list represents the number `999 * 2 = 1998`.

# Constraints
- The number of nodes in the list is in the range $[1, 10^4]$.
- $0 \leq \text{Node.val} \leq 9$.
- The input is generated such that the list represents a number that does not have leading zeros, except the number 0 itself.


```python
from typing import List, Optional
```

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python

class Solution:
    def doubleIt(self, head: Optional[ListNode]) -> Optional[ListNode]:

        if not head:
            return head

        dummy = ListNode(0)
        dummy.next = head

        num   = 0
        count = 0

        while head:
            num    = num * 10 + head.val
            head   = head.next
            count += 1

        ten_power = 10 ** count

        num2 = num * 2
        flag = num2 // (ten_power)

        if flag >= 1:
            
            new_head      = ListNode(1)
            new_head.next = dummy.next

            head       = dummy.next

            num2_digit = num2 // (10 ** count)
            num2       = num2 % (10 ** count)
            count     -= 1

            while head:
                num2_digit = num2 // (10 ** count)
                num2       = num2 % (10 ** count)
                count     -= 1
                head.val   = num2_digit
                head       = head.next

            dummy.next = new_head
            
            return dummy.next

        else:
            
            
            count -= 1
            head   = dummy.next

            while head:
                num2_digit = num2 // (10 ** count)
                num2       = num2  % (10 ** count)
                count     -= 1
                head.val   = num2_digit
                head       = head.next

            return dummy.next
```
