# Problem

# 2 Add Two Numbers

You are given two non-empty linked lists representing two non-negative integers. The digits are stored in reverse order, and each of their nodes contains a single digit. Add the two numbers and return the sum as a linked list.

You may assume the two numbers do not contain any leading zero, except the number `0` itself.

# Examples

**Example 1:**

```
l1 = [2,4,3], l2 = [5,6,4]
```
Output:
```
[7,0,8]
```
Explanation:  

`342 + 465 = 807`

  


**Example 2:**

Input:
```
l1 = [9,9,9,9,9,9,9], l2 = [9,9,9,9]
```
Output:
```
[8,9,9,9,0,0,0,1]
```


# Constraints


    The number of nodes in each linked list is in the range [1, 100].
    0 <= Node.val <= 9
    It is guaranteed that the list represents a number that does not have leading zeros.

````python
def get_digit(l1):

    if l1 is None:
        return None
    else:
        return l1.val

def increment_node(l1):

    if l1 is None:
        return None
    else:
        return l1.next

def get_next_node(l1):

    if l1 is None:
        return None
    else:
        return l1.next

def get_val_node(l1):

    if l1 is None:
        d1_d = 0
    else:
        d1_d = l1.val

    return d1_d

def get_addition_individual_nodes(l1, l2, carry_over):

    d1, d2 = get_val_node(l1), get_val_node(l2)

    d3 = d1 + d2 + carry_over

    d3_01 = d3 % 10
    d3_02 = d3 // 10

    return d3_01, d3_02


````
