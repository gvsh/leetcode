# Problem

# 74. Search a 2D Matrix

# Description

You are given an $m \times n\$ integer matrix `matrix` with the following two properties:

1. Each row is sorted in non-decreasing order.
2. The first integer of each row is greater than the last integer of the previous row.

Given an integer `target`, return `true` if `target` is in `matrix` or `false` otherwise.

You must write a solution in $O(\log(m \times n))$ time complexity.

# Examples

## Example 1

**Input:**  
`matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]], target = 3`  

**Output:**  
`true`

## Example 2

**Input:**  
`matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]], target = 13`  

**Output:**  
`false`

# Constraints

- $ m = \text{matrix.length}$ 
- $ n = \text{matrix[i].length}$ 
- $ 1 \leq m, n \leq 100$ 
- $ -10^4 \leq \text{matrix[i][j]}, \text{target} \leq 10^4$

# Approach

1. First find the row number to search
2. Use binary search for this, if there are 100 rows, search if it's greater than 50th row's first element or smaller.
3. Once you get the correct row number, you search within a row.

```python
def find_row(matrix, target):
    
    m = len(matrix)
    if m > 0:
        n = len(matrix[0])
    
    left, right = 0, m - 1

    while left < right:
    
        print(f"Case 0: {left=}")
        print(f"Case 0: {right=}")
        
        mid = (left + right) // 2
        
        # Check if the target is at the middle
        if matrix[mid][0] == target:
            print(f"Case 1: {left=}")
            print(f"Case 1: {right=}")
            return True
        # If target is smaller than mid element, narrow the search to the left half
        elif matrix[mid][0] > target:
            right = mid - 1
            print(f"Case 2: {left=}")
            print(f"Case 2: {right=}")
        # If target is larger than mid element, narrow the search to the right half
        else:
            left = mid + 1
            print(f"Case 3: {left=}")
            print(f"Case 3: {right=}")

        print("=" * 80)
    
    print(f"{left=}")
    print(f"{right=}")
    print(f"{mid=}")
    
    return -1
```

```python
matrix = [
    [ 1,  2,  3,  4],
    [ 5,  6,  7,  8],
    [10, 11, 16, 20],
    [23, 30, 34, 60]
]
```

```python
target = 6
```

```python
find_row(matrix, target)
```
