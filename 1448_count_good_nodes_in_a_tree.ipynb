# 1448. Count Good Nodes in Binary Tree

# Description
Given a binary tree `root`, a node `X` in the tree is named **good** if, on the path from the root to `X`, there are no nodes with a value greater than `X`.  
Return the number of good nodes in the binary tree.

# Examples

## Example 1

**Input:** `root = [3,1,4,3,null,1,5]`  
**Output:** `4`  
**Explanation:** Nodes in blue are good.  
- Root Node (3) is always a good node.  
- Node 4 → path `(3,4)` has maximum value `4` at the end.  
- Node 5 → path `(3,4,5)` has maximum value `5` at the end.  
- Node 3 → path `(3,1,3)` has maximum value `3` at the end.

## Example 2

**Input:** `root = [3,3,null,4,2]`  
**Output:** `3`  
**Explanation:** Node `2` → path `(3, 3, 2)` is **not** good because `3` (earlier on the path) is greater than `2`.

## Example 3

**Input:** `root = [1]`  
**Output:** `1`  
**Explanation:** Root is considered good.

# Constraints

- The number of nodes in the binary tree is in the range `[1, 10^5]`.
- Each node's value is between `[-10^4, 10^4]`.

```python
# Definition for a binary tree node.

class TreeNode:

    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

    def print_tree(self):
        
        output_string = ""
        output_string += f"\nroot.val: {self.val}\n"
        
        print(f"root.val: {self.val}")

        if self.left:
            print(f"left tree of {self.val}:")
            output_string += f"\nleft tree of {self.val}:\n"
            output_string += self.left.print_tree()

        print(f"Done printing left tree of {self.val}")
        output_string += f"\nDone printing left tree of {self.val}\n"
        print("-" * 40)

        if self.right:
            print(f"right tree of {self.val}:")
            output_string += f"\nright tree of {self.val}:\n"
            output_string += self.right.print_tree()

        print(f"Done printing right tree of {self.val}")
        output_string += f"\nDone printing right tree of {self.val}\n"
        print("-" * 40)


        print(f"Exiting call for {self.val}")
        output_string += f"\nExiting call for {self.val}\n"
        print("=" * 80)
        print("=" * 80)
        
        return output_string

    

    def __str__(self):
        return self.print_tree()
```

```python
def create_tree():
    
    root = TreeNode(3)
    root.left = TreeNode(9)
    root.right = TreeNode(20)
    root.right.left = TreeNode(15)
    root.right.right = TreeNode(7)
    
    return root
```

```python
def print_tree_formatted(node, level=0):
    
    if node:
        print_tree_formatted(node.right, level + 1)
        print(' ' * 4 * level + str(node.val))
        print_tree_formatted(node.left, level + 1)
```

```python
print_tree_formatted(create_tree())
```

```python
preorder = [3, 9, 20, 15, 7]
inorder  = [9, 3, 15, 20, 7]
```

```python
def build_tree(preorder, inorder):

    if not preorder or not inorder:
        return None

    root_val = preorder[0]
    root     = TreeNode(val=root_val)

    root_index = inorder.index(root_val)
    
    left_sub_tree_inorder   = inorder[:root_index]
    right_sub_tree_inorder  = inorder[1 + root_index:]
    
    left_sub_tree_preorder  = preorder[1: len(left_sub_tree_inorder) + 1]
    right_sub_tree_preorder = preorder[1 + len(left_sub_tree_inorder):]

    root.left  = build_tree(left_sub_tree_preorder, left_sub_tree_inorder)
    root.right = build_tree(right_sub_tree_preorder, right_sub_tree_inorder)

    return root
```

```python
sample_tree = build_tree(preorder, inorder)
```

```python
class Solution:

    def count_good_nodes(self, root: TreeNode) -> int:

        good_node_count = 0
        
        def traverse_tree_dfs(node, max_so_far):

            nonlocal good_node_count

            if node.val >= max_so_far:
                good_node_count += 1
                print(node.val)
                max_so_far = max(node.val, max_so_far)
                
            if node.left:
                traverse_tree_dfs(node.left, max_so_far)

            if node.right:
                traverse_tree_dfs(node.right, max_so_far)

        
        if root:
            traverse_tree_dfs(root, -math.inf)

        return good_node_count
        
```

Identify all nodes for which all the nodes in the path are less than the value of the node.
Maximum seen so far
If every node knows the maximum seen so far, than you can determine whether a node is a good node.
If you know the maximum known so far, you can calculate the next maximum known so far.

 
1. When visiting any node we know the parent

2. There is a function which takes a node and the current maximum seen so far before visiting the node.

3. It gives the verdict for the node

```python
class Solution:

    def maxDepth(self, root: TreeNode) -> int:
        if root is None:
            return 0
        else:
            left_height  = self.maxDepth(root.left)
            right_height = self.maxDepth(root.right)
            
            return max(left_height, right_height) + 1
        
```
