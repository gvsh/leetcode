# Problem

# 14 Longest Common Prefix

Write a function to find the longest common prefix string amongst an array of strings.

If there is no common prefix, return an empty string `""`.

# Examples

**Example 1:**

```
strs = ["flower","flow","flight"]
```
Output:
```
"fl"
```

  


**Example 2:**

Input:
```
strs = ["dog","racecar","car"]
```
Output:
```
""
```
Explanation:  
There is no common prefix


# Constraints




`1 <= strs.length <= 200`  
`0 <= strs[i].length <= 200`  
`strs[i]` consists of only lowercase English letters.  




# Solution

````python
import random
from random import choice
from typing import List
````

````python
import string

print(string.ascii_letters)
````

````python
ascii_letters = string.ascii_letters
````

````python
def get_random_lower_case_word(max_length_of_word):

    s_length = random.randint(0, max_length_of_word)
    s        = "".join([choice(ascii_letters) for _ in range(s_length)])

    # return s.lower()
    return ("abcd" + s.lower())
````

````python
def get_random_word():

    s_length = random.randint(0, 15)
    s        = "".join([choice(ascii_letters) for _ in range(s_length)])

    return s
````

````python
single_random_word = get_random_lower_case_word(15)
print(single_random_word)
````

````python
def get_random_words():

    number_of_words     = random.randint(1, 20)

    complete_string_list = []

    for x in range(number_of_words):

        random_word = get_random_lower_case_word(20)

        complete_string_list.append(random_word)


    return complete_string_list
````

````python
random_word_string = get_random_words()
````

````python
from typing import List
import pdb
class Solution:
    def longestCommonPrefix(self, strs: List[str]) -> str:

        longest_common_prefix   = []
        current_char_to_compare = ""
        current_index           = -1
        exit_loop               = 0

        # pdb.set_trace()
        if len(strs) > 0:
            while (exit_loop < 1):
                current_index += 1
                if (current_index <= (len(strs[0]) - 1)):
                    current_char_to_compare = strs[0][current_index]
                else:
                    exit_loop = 1

                exit_for_loop = 0
                for str in strs[1:]:
                    if (current_index > (len(str) - 1)) or (str[current_index] != current_char_to_compare):
                        exit_for_loop = 1

                if (exit_for_loop == 0) and ((exit_loop == 0)):
                    longest_common_prefix.append(current_char_to_compare)
                else:
                    exit_loop = 1


        return "".join(longest_common_prefix)
````

````python
solution_object = Solution()
string_outputs      = solution_object.longestCommonPrefix(random_word_string)
# string_outputs      = solution_object.longestCommonPrefix(["a"])

print(f"{string_outputs=}")

````
