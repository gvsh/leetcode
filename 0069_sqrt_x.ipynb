# Problem 69: Sqrt(x)

# Description
Given a non-negative integer `x`, return the square root of `x` rounded down to the nearest integer. The returned integer should be non-negative as well.

You must not use any built-in exponent function or operator.

For example, do not use `pow(x, 0.5)` in C++ or `x ** 0.5` in Python.

# Examples

## Example 1
**Input:**  
`x = 4`  

**Output:**  
`2`  

**Explanation:**  
The square root of 4 is 2, so we return 2.

## Example 2
**Input:**  
`x = 8`  

**Output:**  
`2`  

**Explanation:**  
The square root of 8 is 2.82842..., and since we round it down to the nearest integer, 2 is returned.

# Constraints
- `0 <= x <= 2^31 - 1`

# Final Submitted

```python

def mySqrt(self, x: int) -> int:
    
    n = x

    if n == 1:
        return 1

    left, right = 0, n

    # mid = (left + right) // 2

    counter = 0
    previous_right = -1
    previous_left  = -1
    previous_final = -1
    final = -1

    # either left changes or right changes or final changes
    # while left <= right:
    while (previous_left != left) or (previous_right != right) or (previous_final != final):

        mid = (left + right) // 2
        mid_square = mid * mid
        
        # print(f"{left=}")
        # print(f"{right=}")
        # print(f"{mid=}")
        # print(f"{mid_square=}")
        

        if mid_square == n:
            
            previous_left = left
            previous_right = right
            previous_final = final
            
            final = mid
            counter += 1
            return final
            # print(f"{previous_final=}")
            # print(f"{final=}")
            
        elif mid_square < n:
            previous_left = left
            previous_right = right
            previous_final = final
            left = mid
            # print(f"{previous_left=}")
            # print(f"{left=}")
            counter += 1
        else:
            previous_left = left
            previous_right = right
            previous_final = final
            right = mid
            # print(f"{previous_right=}")
            # print(f"{right=}")
            counter += 1
        

        # print("=" * 80)
        
        # print(f"{previous_final=}")
        # print(f"{final=}")

        # print(f"{previous_right=}")
        # print(f"{right=}")
        # print(f"{previous_left=}")
        # print(f"{left=}")
        
        # print(f"{final=}")
        
        # print("=" * 80)
        # print("=" * 80)
        # print("=" * 80)
    
    return left
```
