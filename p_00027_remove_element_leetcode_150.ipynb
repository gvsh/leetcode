# Problem

# 88 Merge Sorted Array

Given an integer array `nums` and an integer `val`, remove all occurrences of `val` in `nums` in-place. The order of the elements may be changed. Then return the number of elements in `nums` which are not equal to `val`.

Consider the number of elements in `nums` which are not equal to val be `k`, to get accepted, you need to do the following things:

- Change the array `nums` such that the first `k` elements of `nums` contain the elements which are not equal to `val`. The remaining elements of nums are not important as well as the size of `nums`.
- Return `k`.

**Custom Judge:**

The judge will test your solution with the following code:
```
int[] nums = [...]; // Input array
int val = ...; // Value to remove
int[] expectedNums = [...]; // The expected answer with correct length.
                            // It is sorted with no values equaling val.

int k = removeElement(nums, val); // Calls your implementation

assert k == expectedNums.length;
sort(nums, 0, k); // Sort the first k elements of nums
for (int i = 0; i < actualLength; i++) {
    assert nums[i] == expectedNums[i];
}
```
If all assertions pass, then your solution will be accepted.

# Examples

Input:
```
nums = [3,2,2,3], val = 3
```

```
2, nums = [2,2,_,_]
```
Explanation:  
Your function should return `k = 2`, with the first two elements of `nums` being `2`.
It does not matter what you leave beyond the returned `k` (hence they are underscores).  
  
  


**Example 2:**

```

```

Explanation:  
Your function should return k = 5, with the first five elements of nums containing 0, 0, 1, 3, and 4.
Note that the five elements can be returned in any order.
It does not matter what you leave beyond the returned k (hence they are underscores).

# Approach

1. Background:
    1. `nums` has multiple values of `val`
    2. We scroll through `nums`, whenever we find `val` we replace it with a non `val` value.
    3. When scrolling, each instance of `val`, will be replaced as soon as it is seen.
    4. If we see `val` at the first step, we scroll to find a non val.
    As sson as we find a non val, after finding val, we overwrite val and move non val pointer to the right.
2. Maintain two pointers
2. One is the insert_at_index, this pointer is the location  

The invariants are:  
The val index is always to the left of non val.  
If you're at val, there's no val value left to the left of current val index.
Non val is incremented

1. If you're in a state where you have found val, then you need to put that index in queue, if you find another val while incrementing put that in queue as well.  

Otherwise, if you find non val, you replace val index with non val, delete from queue,

## From Leetcode

1. If `val` is not found at the current index, we copy `nums[j]` to `nums[i]` and increment both indices.
2. Otherwise we increment only `j`.

````python
def maintain_right_pointer_at_non_target_value(signal_to_exit,
                                               right_pointer_index,
                                               count_of_targets_seen_so_far,
                                               target,
                                               left_pointer_index,
                                               nums):

    # Check if right pointer is at target
    # If Yes, try to increment it
    #      If successful in incrementing it, return new right pointer index
    #      Otherwise signal to exit
    # If no, do nothing

    if signal_to_exit == False:

        if nums[right_pointer_index] == target:
            count_of_targets_seen_so_far += 1
            if (right_pointer_index - 1) == left_pointer_index:
                signal_to_exit = True
                return (
                    signal_to_exit,
                    right_pointer_index,
                    count_of_targets_seen_so_far,
                    target,
                    left_pointer_index,
                    nums
                )
            else:
                right_pointer_index -= 1
                (
                    signal_to_exit,
                    right_pointer_index,
                    count_of_targets_seen_so_far,
                    target,
                    left_pointer_index,
                    nums
                ) = maintain_right_pointer_at_non_target_value(
                                                            signal_to_exit=signal_to_exit,
                                                            right_pointer_index=right_pointer_index,
                                                            count_of_targets_seen_so_far=count_of_targets_seen_so_far,
                                                            target=target,
                                                            left_pointer_index=left_pointer_index,
                                                            nums=nums)
        else:
            return (
                    signal_to_exit,
                    right_pointer_index,
                    count_of_targets_seen_so_far,
                    target,
                    left_pointer_index,
                    nums
                )

        return (signal_to_exit,
                right_pointer_index,
                count_of_targets_seen_so_far,
                target,
                left_pointer_index,
                nums)
````

````python
def swap_array_elements(nums, i, j):

    temp1   = nums[i]
    temp2   = nums[j]
    nums[i] = temp2
    nums[j] = temp1

    return nums
````

````python
def increment_left_index_or_swap(signal_to_exit,
                                 right_pointer_index,
                                 count_of_targets_seen_so_far,
                                 target,
                                 left_pointer_index,
                                 nums
                                 ):


    signal_to_exit = False
    swap_flag      = False

    if nums[left_pointer_index] == target:
        count_of_targets_seen_so_far += 1
        if (left_pointer_index + 1) == right_pointer_index:
            signal_to_exit = True
            return (signal_to_exit,
                    right_pointer_index,
                    count_of_targets_seen_so_far,
                    target,
                    left_pointer_index,
                    nums)
        else:
            nums = swap_array_elements(nums,
                                       left_pointer_index,
                                       right_pointer_index)

            left_pointer_index += 1
            swap_flag = True
            (
                    signal_to_exit,
                    right_pointer_index,
                    count_of_targets_seen_so_far,
                    target,
                    left_pointer_index,
                    nums
            ) = maintain_right_pointer_at_non_target_value(
                                                            signal_to_exit=signal_to_exit,
                                                            right_pointer_index=right_pointer_index,
                                                            count_of_targets_seen_so_far=count_of_targets_seen_so_far,
                                                            target=target,
                                                            left_pointer_index=left_pointer_index,
                                                            nums=nums)
    else:
        (
            signal_to_exit,
            right_pointer_index,
            count_of_targets_seen_so_far,
            target,
            left_pointer_index,
            nums
        ) = increment_left_index_or_swap(signal_to_exit,
                                 right_pointer_index,
                                 count_of_targets_seen_so_far,
                                 target,
                                 left_pointer_index,
                                 nums
                                 )

    return (signal_to_exit,
            right_pointer_index,
            count_of_targets_seen_so_far,
            target,
            left_pointer_index,
            nums)
````

````python
def control_flow_for_remove_element(target,
                                    nums):

    right_pointer_index          = len(nums) - 1
    count_of_targets_seen_so_far = 0
    left_pointer_index           = 0

    (
        signal_to_exit,
        right_pointer_index,
        count_of_targets_seen_so_far,
        target,
        left_pointer_index,
        nums
    ) = maintain_right_pointer_at_non_target_value(right_pointer_index=right_pointer_index,
                                                    count_of_targets_seen_so_far=count_of_targets_seen_so_far,
                                                    target=target,
                                                    left_pointer_index=left_pointer_index,
                                                    nums=nums)

    if signal_to_exit:
        return count_of_targets_seen_so_far, nums
    else:
        while True:

            if signal_to_exit:
                return count_of_targets_seen_so_far, nums

            (
                    signal_to_exit,
                    right_pointer_index,
                    count_of_targets_seen_so_far,
                    target,
                    left_pointer_index,
                    nums
            ) = increment_left_index_or_swap(right_pointer_index=(len(nums) - 1),
                                                            count_of_targets_seen_so_far=0,
                                                            target=val,
                                                            left_pointer_index=0,
                                                            nums=nums)


    return count_of_targets_seen_so_far, nums
````
