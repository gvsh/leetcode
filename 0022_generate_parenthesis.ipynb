# Problem 22: Generate Parentheses

# Description
Given `n` pairs of parentheses, write a function to generate all combinations of well-formed parentheses.

# Examples

## Example 1
**Input:**  
`n = 3`  
**Output:**  
`["((()))","(()())","(())()","()(())","()()()"]`

## Example 2
**Input:**  
`n = 1`  
**Output:**  
`["()"]`

# Constraints
- 1 <= n <= 8


```python
from typing import List
```

```python
class Solution:
    def generateParenthesis(self, n: int) -> List[str]:

        paren_list = []


        def is_solution(a, left_count, right_count, n):
            
            return len(a) == 2 * n


        def process_solution(a, left_count, right_count, n):

            paren_list.append(a)


        # def make_move(a, nums):
        #     pass

        # def unmake_move(a, nums):
        #     pass

        def generate_candidates(a, left_count, right_count, n):
            
            candidates = []

            if left_count < n:
                candidates.append([a + "(", left_count + 1, right_count])
            if (left_count > right_count) and right_count < n:
                candidates.append([a + ")", left_count, right_count + 1])

            return candidates

        def backtrack(a, left_count, right_count, n):

            if is_solution(a, left_count, right_count, n):
                
                process_solution(a, left_count, right_count, n)

            else:

                candidates = generate_candidates(a, left_count, right_count, n)

                if candidates:
                    for [candidate, left_count, right_count] in candidates:

                        # make_move(a, nums)

                        backtrack(candidate, left_count, right_count, n)

                        # unmake_move(a, nums)

        left_count = right_count = 0

        backtrack("", left_count, right_count, n)

        return paren_list
```

```python
solution = Solution()
```
