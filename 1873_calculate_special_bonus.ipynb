# Problem 1873: Calculate Special Bonus

## Description
You are given a table `Employees` with the following structure:

| Column Name  | Type    |
|--------------|---------|
| employee_id  | int     |
| name         | varchar |
| salary       | int     |

- `employee_id` is the primary key for this table.
- Each row in the table represents an employee's ID, name, and salary.

Your task is to calculate the bonus for each employee. The bonus is determined as follows:
- The bonus is 100% of the employee's salary if:
  - The `employee_id` is an **odd number**.
  - The `name` does **not** start with the character `'M'`.
- Otherwise, the bonus is `0`.

The result table should be ordered by `employee_id`.

## Examples

### Example 1

**Input:**  
`Employees` table:

| employee_id | name    | salary |
|-------------|---------|--------|
| 2           | Meir    | 3000   |
| 3           | Michael | 3800   |
| 7           | Addilyn | 7400   |
| 8           | Juan    | 6100   |
| 9           | Kannon  | 7700   |

**Output:**  

| employee_id | bonus |
|-------------|-------|
| 2           | 0     |
| 3           | 0     |
| 7           | 7400  |
| 8           | 0     |
| 9           | 7700  |

**Explanation:**
- Employees with IDs 2 and 8 get `0` bonus because their `employee_id` is even.
- The employee with ID 3 gets `0` bonus because their `name` starts with `'M'`.
- The remaining employees (IDs 7 and 9) get a bonus equal to 100% of their salary.

## Constraints
- The `employee_id` is unique.
- The `name` is not null and contains only alphabetic characters.
- The `salary` is a positive integer.

```python
n = 50
```

```python
name_list = ["".join([random.choice(ascii_letters) for _ in range(random.randint(5, 15))]) for _ in range(n)]
```

```python
employee_id_list = list(range(n))
```

```python
salary_list = [random.randint(3, 10) * 10000 for _ in range(n)]
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(zip(employee_id_list, name_list, salary_list),
                  columns=['employee_id', 'name', 'salary'])
```

```python
df.head()
```

```python
no_bonus_df = df[ df['name'].str.lower().str.startswith('m') | df['employee_id'] % 2 == 0]
```

```python
bonus_df    = df[~df['name'].str.lower().str.startswith('m') & df['employee_id'] % 2 == 1]
```
