# Problem 19: Remove Nth Node From End of List

# Description
Given the head of a linked list, remove the nth node from the end of the list and return its head.

# Examples

## Example 1
**Input:**  
`head = [1,2,3,4,5]`, `n = 2`  
**Output:**  
`[1,2,3,5]`

## Example 2
**Input:**  
`head = [1]`, `n = 1`  
**Output:**  
`[]`

## Example 3
**Input:**  
`head = [1,2]`, `n = 1`  
**Output:**  
`[1]`

# Constraints
- The number of nodes in the list is `sz`.
- `1 <= sz <= 30`
- `0 <= Node.val <= 100`
- `1 <= n <= sz`

```python

def removeNthFromEnd(head: Optional[ListNode], n: int) -> Optional[ListNode]:

    if head is None:
        return None

    if head.next is None and n == 1:
        return None

    counter = 0

    dummy   = ListNode(0)
    dummy.next = head

    lagging = dummy
    ahead   = dummy

    while ahead:

        if counter < n + 1:
            ahead = ahead.next
            # print(ahead.val)
            counter += 1
        else:
            ahead   = ahead.next
            lagging = lagging.next
            counter += 1

    print(f"Lagging Node: {lagging.val}")
    
    lagging.next = lagging.next.next
    
    if dummy == lagging.next:
        return lagging


    return head


```
