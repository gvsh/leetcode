# Problem

# 55. Jump Game

**Medium**

---

You are given an integer array `nums`. You are initially positioned at the array's first index, and each element in the array represents your maximum jump length at that position.

Return `true` if you can reach the last index, or `false` otherwise.

---

# Examples
## Example 1

**Input:**  
`nums = [2,3,1,1,4]`  

**Output:**  
`true`  

**Explanation:**  
Jump 1 step from index 0 to 1, then 3 steps to the last index.

---

## Example 2

**Input:**  
`nums = [3,2,1,0,4]`  

**Output:**  
`false`  

**Explanation:**  
You will always arrive at index 3 no matter what. Its maximum jump length is 0, which makes it impossible to reach the last index.

---

# Constraints

- $1 \leq nums.length \leq 10^4$  
- $0 \leq nums[i] \leq 10^5$

```python
nums = [2, 3, 1, 1, 4]
nums = [3, 2, 1, 0, 4]
nums = [1, 1, 1, 0, 0, 0, 2]
```

# Approach

1. At each index, we maintain the farthest distance we can go.
2. At starting index, the initial farthest distance is 0 and we add the nums[0] element to arrive at the new farthest distance.
3. At second index, we update the farthest distance by taking the maximum of `farthest - 1` and `current_element`

```python
def calculate_farthest(nums):
        
    farthest = nums[0]
    
    for num in nums[1:]:
        print(f"{num=}")
        print(f"{farthest=}")
        farthest = max(farthest, num)
        print(f"{farthest=}")
        if farthest >= (len(nums) - 1):
            print(f"{farthest=}")
            return True
        print("=" * 80)
    
    return False
        
        
```

```python
nums = [2, 3, 1, 1, 4]
calculate_farthest(nums)
```
