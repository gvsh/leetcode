# Problem 445: Add Two Numbers II

## Description
You are given two non-empty linked lists representing two non-negative integers. The most significant digit comes first, and each of their nodes contains a single digit. Add the two numbers and return the sum as a linked list.

You may assume the two numbers do not contain any leading zero, except the number 0 itself.

## Examples

### Example 1
**Input:**  
l1 = [7,2,4,3], l2 = [5,6,4]  
**Output:**  
[7,8,0,7]

### Example 2
**Input:**  
l1 = [2,4,3], l2 = [5,6,4]  
**Output:**  
[8,0,7]

### Example 3
**Input:**  
l1 = [0], l2 = [0]  
**Output:**  
[0]

## Constraints
- The number of nodes in each linked list is in the range [1, 100].
- \( 0 \leq \text{Node.val} \leq 9 \)
- It is guaranteed that the list represents a number that does not have leading zeros.

## Follow Up
Could you solve it without reversing the input lists?


```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        
        def reverse_list(in_list):

            prev = None
            curr = in_list

            while curr:

                next = curr.next
                curr.next = prev
                prev = curr
                curr = next

            return prev


        rl1 = reverse_list(l1)
        rl2 = reverse_list(l2)



        digits = []
        carry_over = 0

        l1_head = rl1
        l2_head = rl2


        if (l1_head is None) and (l2_head is None):
            return []

        if l1_head:
            l1_digit = l1_head.val
        else:
            l1_digit = 0

        if l2_head:
            l2_digit = l2_head.val
        else:
            l2_digit = 0

        new_number = l1_digit + l2_digit + carry_over

        new_digit  = new_number %  10
        carry_over = new_number // 10

        l3         = ListNode(val=0, next=None)
        dummy_node = ListNode(val=0, next=l3)

        l3.val = new_digit 

        if l1_head:
            l1_head = l1_head.next
        else:
            l1_head = None
        if l2_head:
            l2_head = l2_head.next
        else:
            l2_head = None



        while l1_head or l2_head:

            if l1_head:
                l1_digit = l1_head.val
            else:
                l1_digit = 0

            if l2_head:
                l2_digit = l2_head.val
            else:
                l2_digit = 0

            new_number = l1_digit + l2_digit + carry_over

            new_digit  = new_number %  10
            carry_over = new_number // 10

            new_l3 = ListNode(val=new_digit, next=None)
            l3.next = new_l3            
            l3 = l3.next

            if l1_head:
                l1_head = l1_head.next
            else:
                l1_head = None
            if l2_head:
                l2_head = l2_head.next
            else:
                l2_head = None
        
        if carry_over > 0:
            new_l3 = ListNode(val=carry_over, next=None)
            l3.next = new_l3

        return reverse_list(dummy_node.next)
```
