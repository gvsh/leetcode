# 199. Binary Tree Right Side View

# Description

Given the root of a binary tree, imagine yourself standing on the right side of it.  
Return the values of the nodes you can see, ordered from top to bottom.

# Examples

## Example 1

Input: root = [1,2,3,null,5,null,4]  
Output: [1,3,4]  
Explanation: From the right side, we see the nodes 1 → 3 → 4.

## Example 2

Input: root = [1,2,3,4,null,null,null,5]  
Output: [1,3,4,5]  
Explanation: From the right side, we see the nodes 1 → 3 → 4 → 5.

## Example 3

Input: root = [1,null,3]  
Output: [1,3]  

## Example 4

Input: root = []  
Output: []  

# Constraints

- The number of nodes in the tree is in the range [0, 100].  
- -100 <= Node.val <= 100

```python
# Definition for a binary tree node.

class TreeNode:

    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

    def print_tree(self):
        
        output_string = ""
        output_string += f"\nroot.val: {self.val}\n"
        
        print(f"root.val: {self.val}")

        if self.left:
            print(f"left tree of {self.val}:")
            output_string += f"\nleft tree of {self.val}:\n"
            output_string += self.left.print_tree()

        print(f"Done printing left tree of {self.val}")
        output_string += f"\nDone printing left tree of {self.val}\n"
        print("-" * 40)

        if self.right:
            print(f"right tree of {self.val}:")
            output_string += f"\nright tree of {self.val}:\n"
            output_string += self.right.print_tree()

        print(f"Done printing right tree of {self.val}")
        output_string += f"\nDone printing right tree of {self.val}\n"
        print("-" * 40)


        print(f"Exiting call for {self.val}")
        output_string += f"\nExiting call for {self.val}\n"
        print("=" * 80)
        print("=" * 80)
        
        return output_string

    

    def __str__(self):
        return self.print_tree()
```

```python
def create_tree():
    
    root = TreeNode(3)
    root.left = TreeNode(9)
    root.right = TreeNode(20)
    root.right.left = TreeNode(15)
    root.right.right = TreeNode(7)
    
    return root
```

```python
def print_tree_formatted(node, level=0):
    
    if node:
        print_tree_formatted(node.right, level + 1)
        print(' ' * 4 * level + str(node.val))
        print_tree_formatted(node.left, level + 1)
```

```python
preorder = [3, 9, 20, 15, 7]
inorder  = [9, 3, 15, 20, 7]
```

```python
def build_tree(preorder, inorder):

    if not preorder or not inorder:
        return None

    root_val = preorder[0]
    root     = TreeNode(val=root_val)

    root_index = inorder.index(root_val)
    
    left_sub_tree_inorder   = inorder[:root_index]
    right_sub_tree_inorder  = inorder[1 + root_index:]
    
    left_sub_tree_preorder  = preorder[1: len(left_sub_tree_inorder) + 1]
    right_sub_tree_preorder = preorder[1 + len(left_sub_tree_inorder):]

    root.left  = build_tree(left_sub_tree_preorder, left_sub_tree_inorder)
    root.right = build_tree(right_sub_tree_preorder, right_sub_tree_inorder)

    return root
```

```python
sample_tree = build_tree(preorder, inorder)
```

```python
print_tree_formatted(sample_tree)
```

```python
def bfs_of_a_tree(root):

    queue = []
    levels = []

    queue.append(root)
    
    while(queue):

        level = []
        
        for i in range(len(queue)):

            current_node = queue.pop(0)

            level.append(current_node)

            if current_node.left:
                queue.append(current_node.left)
            
            if current_node.right:
                queue.append(current_node.right)

        levels.append(level)

    return levels

            
```

```python
def binary_tree_right_side_view(root):

    queue = []
    levels = []

    if root:
        queue.append(root)
    else:
        return None

    while queue:

        level = []

        for i in range(len(queue)):

            current_node = queue.pop(0)

            level.append(current_node)

            if current_node.left:
                queue.append(current_node.left)
                
            if current_node.right:
                queue.append(current_node.right)

        levels.append(level)

    right_side_view = []
    
    for level in levels:
        if level:
            right_side_view.append(level[-1].val)
            
    return right_side_view
```

```python

class Solution:
    def hasPathSum(self, root: Optional[TreeNode], targetSum: int) -> bool:
        
        if not root:
            return False

        targetSum -= root.val

        if not root.left and not root.right:
            return targetSum == 0

        left_path_flag  = False
        right_path_flag = False

        left_path_flag = self.hasPathSum(root.left,  targetSum)

        if not left_path_flag:
            right_path_flag = self.hasPathSum(root.right, targetSum)

        return left_path_flag or right_path_flag
```
