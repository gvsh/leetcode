# Problem

# 58 Length of Last Word

Given a string `s` consisting of words and spaces, return the length of the last word in the string.

A word is a maximal substring consisting of non-space characters only.

# Examples

**Example 1:**




Input:
```
s = "Hello World"
```
Output:
```
5
```
Explanation:  
The last word is "World" with length 5.  
  
  


**Example 2:**

Input:
```
s = "   fly me   to   the moon  "
```
Output:
```
4
```
Explanation:  
The last word is "moon" with length 4.  


# Constraints


- `1 <= s.length <= 104`
- `s` consists of only English letters and spaces `' '`.
- There will be at least one word in `s`.

````python
def calculate_length_of_last_word(s):

    reading_first_word_flag = False

    index_of_first_char = -1
    index_of_last_char  = -1

    for i in range(len(s) - 1, -1, -1):

        current_char = s[i]

        if (current_char == " ") and (reading_first_word_flag == False):
            # print("Case 0")
            pass
        elif (current_char in ascii_letters) and (reading_first_word_flag == False):
            reading_first_word_flag = True
            index_of_first_char     = i
            # print("Case 1")
            # print(f"index_of_first_char {index_of_first_char}")
        elif (current_char == " ") and (reading_first_word_flag == True):
            index_of_last_char      = i
            # print("Case 2")
            # print(f"index_of_last_char {index_of_last_char}")
            return (index_of_first_char - index_of_last_char), s[index_of_last_char + 1: index_of_first_char + 1]
        elif (current_char in ascii_letters) and (reading_first_word_flag == True):
            # print("Case 3 Still Reading")
            pass

    return 0, ""
````

````python
class Solution:
    def lengthOfLastWord(self, s: str) -> int:

        # p is the counter that will be decremented
        # start from the last element of the list
        # Stop at the first element of the start of the word
        last_index      = len(s) - 1
        p               = last_index
        required_length = 0

        while (p >= 0) and (s[p] == " "):
            p -= 1
            # print(p)

        while (p >= 0) and (s[p] != " "):
            p -= 1
            required_length += 1

        return required_length
````
