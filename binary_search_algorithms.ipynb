# Binary Search

```python
from typing import List
```

```python
class Solution:

    def search(self, nums: List[int], target: int) -> int:
        

        def iterative_binary_search(target, left, right):

            nonlocal nums


            while left <= right:
                
                mid = left + (right - left) // 2

                if nums[mid] > target:
                    right = mid - 1
                elif nums[mid] < target:
                    left = mid + 1
                else:
                    return mid
            
            return -1

        return iterative_binary_search(target, 0, len(nums) - 1)
```

```python
class Solution:

    def search(self, nums: List[int], target: int) -> int:
        
        def recursive_binary_search(target, left, right):

            nonlocal nums

            if left > right:
                return -1

            mid = left + (right - left) // 2

            if target > nums[mid]:

                return recursive_binary_search(target, mid + 1, right)
            
            elif target < nums[mid]:

                return recursive_binary_search(target, left, mid - 1)

            else:

                return mid

        return recursive_binary_search(target, 0, len(nums) - 1)
```
