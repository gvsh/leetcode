# Problem

# 205 Isomorphic Strings

Given two strings `s` and `t`, determine if they are isomorphic.

Two strings `s` and `t` are isomorphic if the characters in `s` can be replaced to get `t`.

All occurrences of a character must be replaced with another character while preserving the order of characters. No two characters may map to the same character, but a character may map to itself.

# Examples

**Example 1:**

```
s = "egg", t = "add"
```
Output:
```
true
```
Explanation:  
The strings `s` and `t` can be made identical by:
- Mapping `e` to `a`.
- Mapping `g` to `d`.  
  
  


**Example 2:**

Input:
```
s = "foo", t = "bar"
```
Output:
```
false
```
Explanation:  
The strings `s` and `t` can not be made identical as `o` needs to be mapped to both `a` and `r`.


**Example 3:**

Input:
```
ransomNote = "aa", magazine = "aab"
```
Output:
```
true
```

# Constraints


- $1$ `<= ransomNote.length, magazine.length <= `$10^5$
- `ransomNote` and `magazine` consist of lowercase English letters.


# Solution

````python
import random
from random import choice
from typing import List, Tuple, Union, Optional
````

````python
import string

print(string.ascii_letters)
````

````python
ascii_letters = string.ascii_letters
````

````python
def get_random_lower_case_word(length_of_word):

    s        = "".join([choice(ascii_letters[:10]) for _ in range(length_of_word)])

    return (s.lower())
````

````python
def get_s_and_t_test_suite(number_of_test_cases):

    test_suite = []

    for x in range(number_of_test_cases):

        length_of_word = random.randint(1, 20)

        s = get_random_lower_case_word(length_of_word)
        t = get_random_lower_case_word(length_of_word)

        test_tuple = s, t

        test_suite.append(test_tuple)

    return test_suite
````

````python
test_suite = get_s_and_t_test_suite(50)
````

````python
from collections import defaultdict
from collections import Counter
````

````python
def can_construct(s, t):

    # print(f"{s=}")
    # print(f"{t=}")

    s_counter = Counter(s)
    t_counter = Counter(t)

    # print(f"{s_counter=}")
    # print(f"{t_counter=}")

    s_list = list(s_counter.items())
    t_list = list(t_counter.items())

    # print(f"{s_list=}")
    # print(f"{t_list=}")

    s_list_values = list(map(lambda x: x[1], s_list))
    t_list_values = list(map(lambda x: x[1], t_list))

    s_list_values = sorted(s_list_values, reverse=True)
    t_list_values = sorted(t_list_values, reverse=True)


    s_dict = defaultdict(list)
    t_dict = defaultdict(list)

    for i, c in enumerate(s):
        s_dict[c].append(i)
    for i, c in enumerate(t):
        t_dict[c].append(i)

    print(f"{s_list_values=}")
    print(f"{t_list_values=}")

    print(f"{s_dict=}")
    print(f"{t_dict=}")

    return s_list_values == t_list_values
````
