# Problem

# 28 Find the Index of the First Occurrence in a String


Given two strings `needle` and `haystack`, return the index of the first occurrence of `needle` in `haystack`, or `-1` if `needle` is not part of `haystack`.

# Examples

**Example 1:**

Input:  
```
haystack = "sadbutsad", needle = "sad"
```
Output:
```
0
```
Explanation:  
"sad" occurs at index 0 and 6.
The first occurrence is at index 0, so we return 0.
  


**Example 2:**

Input:
```
haystack = "leetcode", needle = "leeto"
```
Output:
```
-1
```
Explanation:  
"leeto" did not occur in "leetcode", so we return -1.


# Constraints

`1 <= haystack.length, needle.length <= 104`  
`haystack` and `needle` consist of only lowercase English characters.

````python
def get_random_lower_case_word(max_length_of_word):

    s_length = random.randint(0, max_length_of_word)
    s        = "".join([choice(ascii_letters) for _ in range(s_length)])

    # return s.lower()
    return ("abcd" + s.lower())
````

````python
def get_needle_and_haystack(max_length_of_haystack=30):

    haystack_length = random.randint(0, max_length_of_haystack)
    needle_length   = random.randint(0, haystack_length)

    haystack = get_random_lower_case_word(haystack_length)
    needle   = get_random_lower_case_word(needle_length)

    return needle, haystack
````
