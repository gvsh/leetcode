# Problem

# 35. Search Insert Position

# Description

Given a sorted array of distinct integers and a target value, return the index if the target is found. If not, return the index where it would be if it were inserted in order.

You must write an algorithm with O(log n) runtime complexity.

# Examples

## Example 1

**Input:**  
`nums = [1,3,5,6], target = 5`  

**Output:**  
`2`

## Example 2

**Input:**  
`nums = [1,3,5,6], target = 2`  

**Output:**  
`1`

## Example 3

**Input:**  
`nums = [1,3,5,6], target = 7`  

**Output:**  
`4`

# Constraints

- $ 1 \leq \text{nums.length} \leq 10^4$ 
- $ -10^4 \leq \text{nums[i]} \leq 10^4$ 
- `nums` contains distinct values sorted in ascending order.
- $ -10^4 \leq \text{target} \leq 10^4$

```python
def array_binary_search(nums, target, add_index):

    mid       = len(nums) // 2
    mid_value = nums[mid]
    
    print("=" * 80)
    print(f"{nums=}")
    print(f"{target=}")
    print(f"{add_index=}")
    print(f"{mid=}")
    print(f"{mid_value=}")

    
    if target == mid_value:
        return mid + add_index
    if target > nums[-1]:
        return add_index + len(nums)
    elif target < nums[0]:
        return add_index
    elif target < mid_value:
        left_array = nums[0: mid]
        return array_binary_search(left_array, target, add_index)
    elif target > mid_value:
        right_array = nums[mid + 1:]
        return array_binary_search(right_array, target, add_index + mid + 1)
```
