# Problem

# 219 Contains Duplicate

Given an integer array nums and an integer `k`, return `true` if there are two distinct indices `i` and `j` in the array such that `nums[i] == nums[j]` and `abs(i - j) <= k`.

# Examples

**Example 1:**

```
nums = [1,2,3,1], k = 3
```
Output:
```
true
```

  


**Example 2:**

Input:
```
nums = [1,0,1,1], k = 1
```
Output:
```
true
```
  


**Example 3:**

Input:
```
nums = [1,2,3,1,2,3], k = 2
```
Output:
```
false
```


# Constraints


- `1 <= nums.length <= 105`
- `-109 <= nums[i] <= 109`
- `0 <= k <= 105`

````python
def contains_duplicate(nums, k):

    len_nums = len(nums)

    if k < len_nums:
        for i, x in enumerate(nums[:len_nums - k]):
            hash_map = {}
            for j in range(k + 1):
                current_value = nums[i + j]
                current_index = i + j
                print(f"index: {current_index}, value: {current_value}")
                if current_value in hash_map:
                    hash_map[current_value].append(current_index)
                    return True
                else:
                    hash_map[current_value] = [current_index]
    else:
        hash_map = {}
        for i, x in enumerate(nums):
            current_value = nums[i]
            current_index = i
            print(f"index: {current_index}, value: {current_value}")
            if current_value in hash_map:
                hash_map[current_value].append(current_index)
                return True
            else:
                print("Adding")
                hash_map[current_value] = [current_index]


    return False
````
