# Problem 257. Binary Tree Paths

# Description
Given the root of a binary tree, return all root-to-leaf paths in any order.

A leaf is a node with no children.

# Examples

## Example 1
**Input:**  
root = [1,2,3,null,5]  
**Output:**  
["1->2->5","1->3"]

## Example 2
**Input:**  
root = [1]  
**Output:**  
["1"]

# Constraints
- The number of nodes in the tree is in the range [1, 100].
- -100 <= Node.val <= 100


```python
from typing import List, Optional
```

```python

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
# Definition for a binary tree node.
```

```python

class Solution:
    def binaryTreePaths(self, root: Optional[TreeNode]) -> List[str]:
        
        path_list = []

        def is_solution(a, n):

            return (n is not None) and (n.left is None) and (n.right is None)


        def process_solution(a, n):

            path_list.append(a[::])


        def generate_candidates(a, n):

            candidates = []
            
            if n.left:
                candidates.append(n.left)
            if n.right:
                candidates.append(n.right)

            return candidates


        def backtrack(a, n):

            if is_solution(a, n):
                process_solution(a, n)
            else:

                candidates = generate_candidates(a, n)

                for candidate in candidates:

                    # make_move(a, n)

                    backtrack(a + "->" + str(candidate.val), candidate)

                    # unmake_move(a, n)

        if root:
            backtrack(str(root.val), root)
        else:
            return path_list

        return path_list
```
