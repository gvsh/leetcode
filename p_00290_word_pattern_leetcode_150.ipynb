# Problem

# 205 Isomorphic Strings

Given two strings `s` and `t`, determine if they are isomorphic.

Two strings `s` and `t` are isomorphic if the characters in `s` can be replaced to get `t`.

All occurrences of a character must be replaced with another character while preserving the order of characters. No two characters may map to the same character, but a character may map to itself.

# Examples

**Example 1:**

```
s = "egg", t = "add"
```
Output:
```
true
```
Explanation:  
The strings `s` and `t` can be made identical by:
- Mapping `e` to `a`.
- Mapping `g` to `d`.  
  
  


**Example 2:**

Input:
```
s = "foo", t = "bar"
```
Output:
```
false
```
Explanation:  
The strings `s` and `t` can not be made identical as `o` needs to be mapped to both `a` and `r`.


**Example 3:**

Input:
```
ransomNote = "aa", magazine = "aab"
```
Output:
```
true
```

````python
from collections import defaultdict
from collections import Counter
````

````python
def get_random_lower_case_word(max_number_of_unique_characters, length_of_word):

    s        = "".join([choice(ascii_letters[:max_number_of_unique_characters]) for _ in range(length_of_word)])

    return (s.lower())
````

````python
def get_random_sentence(first_n_words, number_of_words):

    random_shift = random.randint(1, 25)

    random_words_list = [random.choice(common_words[1 + random_shift: first_n_words + random_shift]) for _ in range(number_of_words)]

    return " ".join(random_words_list)
````

````python
def get_pattern_and_s_test_suite(number_of_test_cases):

    test_suite = []

    for x in range(number_of_test_cases):

        max_number_of_unique_characters = 5
        length_of_word = random.randint(1, 6)

        pattern = get_random_lower_case_word(max_number_of_unique_characters, length_of_word)
        s = get_random_sentence(max_number_of_unique_characters, length_of_word)

        test_tuple = pattern, s

        test_suite.append(test_tuple)

    return test_suite
````

````python
test_suite = get_pattern_and_s_test_suite(30)
````

````python
def can_construct(pattern, s):

    print(f"{pattern=}")

    word_list = s.split()
    print(f"{word_list=}")

    if len(word_list) != len(pattern):
        return False

    # char_to_word_map = defaultdict(lambda: "")
    # word_to_char_map = defaultdict(lambda: "")
    char_to_word_map = {}
    word_to_char_map = {}


    for i, current_char in enumerate(pattern):

        current_word = word_list[i]

        if current_char in char_to_word_map:
            if current_word == char_to_word_map[current_char]:
                pass
            else:
                print(f"`{current_char}` should map to `{char_to_word_map[current_char]}` but found `{current_word}` at `{i}` index.")
                return False
        else:
            char_to_word_map[current_char] = current_word


    for i, current_word in enumerate(word_list):

        current_char = pattern[i]

        if current_word in word_to_char_map:
            if current_char == word_to_char_map[current_word]:
                pass
            else:
                print(f"`{current_word}` should map to `{word_to_char_map[current_word]}` but found `{current_char}` at `{i}` index.")
                return False
        else:
            word_to_char_map[current_word] = current_char


    return True
````
