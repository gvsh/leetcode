# Problem 129: Sum Root to Leaf Numbers

# Description
You are given the root of a binary tree containing digits from 0 to 9 only.

Each root-to-leaf path in the tree represents a number.

For example, the root-to-leaf path `1 -> 2 -> 3` represents the number `123`.

Return the total sum of all root-to-leaf numbers. Test cases are generated so that the answer will fit in a 32-bit integer.

A leaf node is a node with no children.

# Examples

## Example 1
**Input:**  
`root = [1,2,3]`  
**Output:**  
`25`  
**Explanation:**  
- The root-to-leaf path `1->2` represents the number `12`.  
- The root-to-leaf path `1->3` represents the number `13`.  
- Therefore, `sum = 12 + 13 = 25`.

## Example 2
**Input:**  
`root = [4,9,0,5,1]`  
**Output:**  
`1026`  
**Explanation:**  
- The root-to-leaf path `4->9->5` represents the number `495`.  
- The root-to-leaf path `4->9->1` represents the number `491`.  
- The root-to-leaf path `4->0` represents the number `40`.  
- Therefore, `sum = 495 + 491 + 40 = 1026`.

# Constraints
- The number of nodes in the tree is in the range `[1, 1000]`.
- `0 <= Node.val <= 9`
- The depth of the tree will not exceed `10`.


```python
from typing import Optional
```

```python
# Definition for a binary tree node.
```

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python

class Solution:
    def sumNumbers(self, root: Optional[TreeNode]) -> int:


        def process_empty_node():

            return 0


        def is_leaf(node):

            if not node.left and not node.right:
                return True


        def calculate_sum_root_to_leaf(node, node_num):
            
            nonlocal total_sum

            if not node:
                return process_empty_node()
            

            if is_leaf(node):

                # print("=" * 80)
                # print(f"{node.val=}")
                # print("=" * 80)

                node_num = node_num * 10 + node.val
                current_sum = node_num

                # print(f"{current_sum=}")

                # print(f"{result.total_sum=}")
                total_sum += current_sum
                # print(f"{total_sum=}")
                
                # print("=" * 80)

            
            else:
                node_num = node_num * 10 + node.val

                if node.left:
                    # left(node, result)
                    calculate_sum_root_to_leaf(node.left, node_num)

                if node.right:
                    calculate_sum_root_to_leaf(node.right, node_num)
                    # right(node, result)

        total_sum = 0
        
        

        calculate_sum_root_to_leaf(root, 0)

        # print(f"{result=}")
        # print(f"{result.total_sum=}")
        # print(f"{result.node_num=}")
        print(f"{total_sum=}")

        return total_sum
```
