# Problem

# 49 Group Anagrams

Given an array of strings `strs`, group the anagrams together.  
You can return the answer in any order.

# Examples

**Example 1:**


Input:
```
strs = ["eat","tea","tan","ate","nat","bat"]
```
Output:
```
[["bat"],["nat","tan"],["ate","eat","tea"]]
```
Explanation:  

- There is no string in strs that can be rearranged to form "bat".
- The strings "nat" and "tan" are anagrams as they can be rearranged to form each other.
- The strings "ate", "eat", and "tea" are anagrams as they can be rearranged to form each other.

  
  


**Example 2:**

Input:
```
strs = ["a"]
```
Output:
```
[["a"]]
```


# Constraints


- `1 <= strs.length <= 104`
- `0 <= strs[i].length <= 100`
- `strs[i]` consists of lowercase English letters.

````python
import string

ascii_printable       = list(string.printable)
ascii_letters_numbers = list(string.ascii_letters + string.digits)


class Solution:
    def get_letter_only_s(self, printable_s: str) -> str:

        letter_s = "".join([x for x in printable_s if x in ascii_letters_numbers])

        return letter_s.lower()


    def isPalindrome(self, s: str) -> bool:

        s_prime = get_letter_only_s(s)

        i = 0
        j = len(s_prime) - 1

        if len(s_prime) > 1:
            while (i < j):
                # print(s_prime[i])
                # print(s_prime[j])
                # print("*********")

                if s_prime[i] == s_prime[j]:
                    i += 1
                    j -= 1
                else:
                    # print("Not Palindrome")
                    return False

        return True


````
