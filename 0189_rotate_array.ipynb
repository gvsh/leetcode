# Problem

Given an integer array `nums`, rotate the array to the right by `k` steps, where `k` is non-negative.

---

# Examples
## Example 1

**Input:**  
`nums = [1,2,3,4,5,6,7], k = 3`  
**Output:**  
`[5,6,7,1,2,3,4]`  

**Explanation:**  
- Rotate 1 step to the right: `[7,1,2,3,4,5,6]`  
- Rotate 2 steps to the right: `[6,7,1,2,3,4,5]`  
- Rotate 3 steps to the right: `[5,6,7,1,2,3,4]`

---

## Example 2

**Input:**  
`nums = [-1,-100,3,99], k = 2`  
**Output:**  
`[3,99,-1,-100]`

**Explanation:**  
- Rotate 1 step to the right: `[99,-1,-100,3]`  
- Rotate 2 steps to the right: `[3,99,-1,-100]`

---

# Constraints

- `1 <= nums.length <= 10^5`  
- `-2^31 <= nums[i] <= 2^31 - 1`  
- `0 <= k <= 10^5`  

---

# Follow-Up

1. Try to come up with as many solutions as you can. There are at least three different ways to solve this problem.  
2. Could you do it in-place with O(1) extra space?

````python
nums = [-1, -100, 3, 99]
k = 2
len_array = len(nums)
si   = 0
seen_indices = []
backup = nums[0]
swap   = nums[0]
counter = 0

while True:
    print("=" * 120)
    print("=" * 120)
    print(f"{counter=}")
    
    ni = get_new_index(si, k, len_array)
    print(f"{counter=}")
    

    print(f"{nums[ni]=}")
    swap     = nums[ni]
    nums[ni] = backup
    backup   = swap
    print(f"{backup=}")
    print(f"{swap=}")
    print(f"{nums[ni]=}")
    seen_indices.append(si)        
    si       = ni
    print(f"{counter=}")
    counter  += 1
    print(f"{seen_indices=}")
    
    if si in seen_indices:
        if len(seen_indices) == len_array:
            break
        else:
            si = (si + 1) % len_array
            backup = nums[si]
            swap   = nums[si]
            
    print("=" * 120)
    print("=" * 120)
````
