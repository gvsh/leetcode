# Problem 50: Pow(x, n)

# Description
Implement `pow(x, n)`, which calculates $ x^n $ (i.e., $ x $ raised to the power $ n $).

# Examples

## Example 1
**Input**:  
x = 2.00000, n = 10  
**Output**:  
1024.00000  

## Example 2
**Input**:  
x = 2.10000, n = 3  
**Output**:  
9.26100  

## Example 3
**Input**:  
x = 2.00000, n = -2  
**Output**:  
0.25000  
**Explanation**:  
$ 2^{-2} = 1 / 2^2 = 1 / 4 = 0.25 $

# Constraints
- $ -100.0 < x < 100.0 $
- $ -2^{31} \leq n \leq 2^{31} - 1 $
- $ n $ is an integer.
- Either $ x \neq 0 $ or $ n > 0 $.
- $ -10^4 \leq x^n \leq 10^4 $

# Follow Up
Consider optimizing the implementation to handle very large values of $ n $ efficiently.

```python
def pow(x, n):
    if n > 0:
        if n % 2 == 0:
            return pow((x * x), n // 2)
        elif n % 2 == 1:
            return x * pow(x * x, (n - 1) // 2)
    elif n < 0:
        return 1 / pow(x, -n)
    elif n == 0:
        return 1
        
    
```
