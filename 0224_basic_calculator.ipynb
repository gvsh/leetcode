# Description

Given a string `s` representing a valid expression, implement a basic calculator to evaluate it, and return the result of the evaluation.

**Note:** You are not allowed to use any built-in function which evaluates strings as mathematical expressions, such as `eval()`.

---

# Examples

## Example 1:
**Input:** 

`s = "1 + 1"`

**Output:** 

`2`

## Example 2:
**Input:** 

`s = " 2-1 + 2 "`

**Output:** 

`3`

## Example 3:
**Input:** 

`s = "(1+(4+5+2)-3)+(6+8)"`

**Output:** 

`23`


---

# Constraints

- $ 1 \leq s.length \leq 3 \times 10^5 $
- `s` consists of digits, `+`, `-`, `(`, `)`, and `' '`.
- `s` represents a valid expression.
- `'+'` is not used as a unary operation (i.e., `"+1"` and `"+(2 + 3)"` are invalid).
- `'-'` **could** be used as a unary operation (i.e., `"-1"` and `"-(2 + 3)"` are valid).
- There will be no two consecutive operators in the input.
- Every number and running calculation will fit in a signed 32-bit integer.

# Approach

1. Get tokens list
2. Solve for the case without brackets

1. If there are no brackets:
2. Find all occurences of / and solve and update the list.
3. Find all occurences of * and solve and update the list.

```python
s = "(1 * (114 + 15 + 2)- 13) * (6 + 8)"
s = "114 + 15 + 2 - 10 / 2 * 6 / 2"
list(s)
```

```python
tokens = []
tokens_char_list = list(s)
non_digits = ["(", ")", "+", "-", "*", "/", " "]

current_digit = ""

for i in range(len(tokens_char_list)):
    
    current_token = tokens_char_list[i]
    
    if current_token in non_digits:
        if len(current_digit) > 0:
            tokens.append(current_digit)
            current_digit = ""
        if current_token != " ":
            tokens.append(current_token)
            current_digit = ""
    else:
        current_digit += current_token
        if i == len(tokens_char_list) - 1:
            tokens.append(current_digit)
            
```

```python
def solve_without_brackets(tokens, operator):
    
    new_tokens = []

    if operator in tokens:
        
        for i in range(len(tokens) - 1):
            if tokens[i] == operator:
                
                if operator == "/":
                    result = int(tokens[i - 1]) / int(tokens[i + 1])
                if operator == "*":
                    result = int(tokens[i - 1]) * int(tokens[i + 1])
                if operator == "-":
                    result = int(tokens[i - 1]) - int(tokens[i + 1])
                if operator == "+":
                    result = int(tokens[i - 1]) + int(tokens[i + 1])
                    print(f"{result=}")
                
                if len(tokens) > (i + 2):
                    print(f"1. {len(tokens)=}")
                    print(f"1. {(i + 2)=}")
                    print(f"1. {tokens=}")
                    new_tokens = tokens[:i - 1] + [result] + tokens[i + 2:]
                    print(f"1. {new_tokens=}")
                else:
                    print(f"2. {len(tokens)=}")
                    print(f"2. {(i + 2)=}")
                    print(f"2. {tokens=}")
                    new_tokens = tokens[:i - 1] + [result]
                    print(f"2. {new_tokens=}")
    
        print(f"3. {new_tokens=}")
        return new_tokens, 0
    else:
        print(f"4. {tokens=}")
        return tokens, 1

flag = 0

print(f"0. {tokens=}")

print("=" * 80)
print("=" * 80)
while True:
    if flag == 1:
        print("Done")
        break
    else:
        tokens, flag = solve_without_brackets(tokens, "/")
print("=" * 80)
print("=" * 80)

flag = 0
while True:
    if flag == 1:
        print("Done")
        break
    else:
        tokens, flag = solve_without_brackets(tokens, "*")

flag = 0
print("=" * 80)
print("=" * 80)
while True:
    if flag == 1:
        print("Done")
        break
    else:
        tokens, flag = solve_without_brackets(tokens, "-")

flag = 0
print("=" * 80)
print("=" * 80)
while True:
    if flag == 1:
        print("Done")
        break
    else:
        tokens, flag = solve_without_brackets(tokens, "+")
print("=" * 80)
print("=" * 80)
        
```
