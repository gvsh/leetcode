# Problem

# 149 Max Points on a Line

Given an array of `points` where `points[i]` = $[x_i, y_i]$ represents a point on the **X-Y** plane, return the maximum number of points that lie on the same straight line.

# Examples

**Example 1:**

```
points = [[1,1],[2,2],[3,3]]
```
Output:
```
3
```

  


**Example 2:**

Input:
```
points = [[1,1],[3,2],[5,3],[4,1],[2,3],[1,4]]
```
Output:
```
4
```



# Constraints




    1 <= points.length <= 300
    points[i].length == 2
    -104 <= xi, yi <= 104
    All the points are unique.

````python
import random
from random import choice
from typing import List
from pydantic import BaseModel
````

````python
class Point(BaseModel):
    x: int
    y: int
````

````python
def get_random_point(max_coordinate_value):

    x_coordinate = random.randint(-max_coordinate_value, max_coordinate_value)
    y_coordinate = random.randint(-max_coordinate_value, max_coordinate_value)
    point = Point(x=x_coordinate, y=y_coordinate)

    return point
````

````python
def get_random_points():

    max_coordinate_value = random.randint(3, 10)
    max_coordinate_value = 3
    n_points             = random.randint(5, 15)
    set_of_points        = [get_random_point(max_coordinate_value) for _ in range(n_points)]

    return set_of_points
````
