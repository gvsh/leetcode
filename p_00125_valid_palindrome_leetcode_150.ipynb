# 88 Merge Sorted Array

A phrase is a palindrome if, after converting all uppercase letters into lowercase letters and removing all non-alphanumeric characters, it reads the same forward and backward.  
Alphanumeric characters include letters and numbers.

Given a string `s`, return `true` if it is a palindrome, or `false` otherwise.

# Examples

**Example 1:**


Input:
```
s = "A man, a plan, a canal: Panama"
```
Output:
```
true
```
Explanation:  
"amanaplanacanalpanama" is a palindrome.   
  
  


**Example 2:**

Input:
```
s = "race a car"
```
Output:
```
false
```
Explanation:  
"raceacar" is not a palindrome.


# Constraints


`1 <= s.length <= 2 * 105`  
`s` consists only of printable ASCII characters.

````python
import string

ascii_printable       = list(string.printable)
ascii_letters_numbers = list(string.ascii_letters + string.digits)


class Solution:
    def get_letter_only_s(self, printable_s: str) -> str:

        letter_s = "".join([x for x in printable_s if x in ascii_letters_numbers])

        return letter_s.lower()


    def isPalindrome(self, s: str) -> bool:

        s_prime = get_letter_only_s(s)

        i = 0
        j = len(s_prime) - 1

        if len(s_prime) > 1:
            while (i < j):
                # print(s_prime[i])
                # print(s_prime[j])
                # print("*********")

                if s_prime[i] == s_prime[j]:
                    i += 1
                    j -= 1
                else:
                    # print("Not Palindrome")
                    return False

        return True


````
