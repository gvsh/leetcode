# 6 Zigzag Conversion

The string `"PAYPALISHIRING"` is written in a zigzag pattern on a given number of rows like this: (you may want to display this pattern in a fixed font for better legibility)

```
P   A   H   N  
A P L S I I G  
Y   I   R  
```

And then read line by line: `"PAHNAPLSIIGYIR"`   

Write the code that will take a string and make this conversion given a number of rows:  

    string convert(string s, int numRows);

````python
def get_ascending_descending(n):

    nn = n - 1

    ascending_sequence  = []
    descending_sequence = []

    ascending_dictionary  = {}
    descending_dictionary = {}


    for ci in range(0, 40):
        if (ci % (nn * 2) // nn) == 1:
            ascending_sequence.append(ci)
        else:
            descending_sequence.append(ci)


    for ci in ascending_sequence:

        dict_value = (ci * -1 ) % nn
        if dict_value == 0:
            dict_value = nn
            ascending_dictionary[ci] = dict_value
        else:
            ascending_dictionary[ci] = dict_value

    for ci in descending_sequence:

        dict_value = ci % nn
        if dict_value == 0:
            # dict_value = nn
            descending_dictionary[ci] = dict_value
        else:
            descending_dictionary[ci] = dict_value

    return ascending_sequence, descending_sequence, ascending_dictionary, descending_dictionary
````

````python

class Solution:
    def get_letter_only_s(self, printable_s: str) -> str:

        letter_s = "".join([x for x in printable_s if x in ascii_letters_numbers])

        return letter_s.lower()


    def isPalindrome(self, s: str) -> bool:

        s_prime = get_letter_only_s(s)

        i = 0
        j = len(s_prime) - 1

        if len(s_prime) > 1:
            while (i < j):
                # print(s_prime[i])
                # print(s_prime[j])
                # print("*********")

                if s_prime[i] == s_prime[j]:
                    i += 1
                    j -= 1
                else:
                    # print("Not Palindrome")
                    return False

        return True


````
