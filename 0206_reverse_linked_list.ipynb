# Problem 206: Reverse Linked List

# Description

Given the head of a singly linked list, reverse the list, and return the reversed list.

# Examples

## Example 1

**Input:**  
`head = [1,2,3,4,5]`  
**Output:**  
`[5,4,3,2,1]`

## Example 2

**Input:**  
`head = [1,2]`  
**Output:**  
`[2,1]`

## Example 3

**Input:**  
`head = []`  
**Output:**  
`[]`

# Constraints

- The number of nodes in the list is in the range `[0, 5000]`.
- `-5000 <= Node.val <= 5000`

## Iterative Solution

1. We iterate over the list and fix the next pointers so that the list is reversed.
2. For three nodes in sequence: n1, n2, n3
3. n1.next = n2
4. n2.next = n3
5. In the output we want, n2.next = n1, n3.next = n2
6. While iterating we keep a backup dummy node.
7. backup = ListNode(n2.val,

1. There's no connection between `prev` and `curr` nodes at the end of each `while` loop iteration.
2. `prev` starts a list that is reversed, `curr` holds the list to be fixed.

# Final Solution

```python
class Solution:
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:

        dummy = ListNode(0)
        dummy.next = head
        
        prev = None
        curr = head

        print(f"{curr.val=}")
        
        while curr:

            next = curr.next
            curr.next = prev
            prev = curr
            curr = next

            print("=" * 80)
            print("printing prev list")
            print_list(prev)
            print("done printing prev list")
            
            print("=" * 80)
            print("printing curr list")
            print_list(curr)
            print("done printing curr list")
            print("=" * 80)
            
            # if prev:
            #     print(f"{prev.val=}")

            #     if prev.next:
            #         print(f"{prev.next.val=}")
            #     # print_list(dummy.next)
            #         if prev.next.next:
            #             print(f"{prev.next.next.val=}")
            #     # print_list(dummy.next)
                
            # if curr:
            #     print(f"{curr.val=}")

            #     if curr.next:
            #         print(f"{curr.next.val=}")
            #     # print_list(dummy.next)

            print("=" * 80)
            print("=" * 80)
            print("=" * 80)

        # print("*" * 80)
        
        return prev
```
