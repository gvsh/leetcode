# Problem 78: Subsets

# Description
Given an integer array `nums` of unique elements, return all possible subsets (the power set).  
The solution set must not contain duplicate subsets. Return the solution in any order.

# Examples

## Example 1
**Input:**  
nums = [1,2,3]  
**Output:**  
[[],[1],[2],[1,2],[3],[1,3],[2,3],[1,2,3]]  

## Example 2
**Input:**  
nums = [0]  
**Output:**  
[[],[0]]  

# Constraints
- 1 <= nums.length <= 10  
- -10 <= nums[i] <= 10  
- All the numbers of `nums` are unique.


```python
from typing import List
```

```python
class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        

        subset_list = []


        def is_solution(a, nums):
            
            return len(a) == len(nums)


        def process_solution(a, nums):

            current_subset = []

            for i, element in enumerate(a):
                if element:
                    current_subset.append(nums[i])
            
            subset_list.append(current_subset)


        # def make_move(a, nums):
        #     pass

        # def unmake_move(a, nums):
        #     pass

        def generate_candidates(a, nums):
            
            if len(a) < len(nums):
                return [True, False]


        def backtrack(a, nums):

            if is_solution(a, nums):
                
                process_solution(a, nums)

            else:

                candidates = generate_candidates(a, nums)

                for candidate in candidates:

                    # make_move(a, nums)

                    backtrack(a + [candidate], nums)

                    # unmake_move(a, nums)


        backtrack([], nums)

        return subset_list
```
