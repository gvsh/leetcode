# Problem 2095. Delete the Middle Node of a Linked List

## Description
You are given the head of a linked list. Delete the middle node, and return the head of the modified linked list.

The middle node of a linked list of size `n` is the ⌊n / 2⌋th node from the start using 0-based indexing, where ⌊x⌋ denotes the largest integer less than or equal to `x`.

- For `n = 1, 2, 3, 4, and 5`, the middle nodes are `0, 1, 1, 2, and 2`, respectively.

## Examples

### Example 1
**Input:**  
`head = [1,3,4,7,1,2,6]`  
**Output:**  
`[1,3,4,1,2,6]`  
**Explanation:**  
The linked list is represented as: `[1,3,4,7,1,2,6]`.  
Since `n = 7`, the middle node (index `3`, value `7`) is removed. The resulting list is `[1,3,4,1,2,6]`.

### Example 2
**Input:**  
`head = [1,2,3,4]`  
**Output:**  
`[1,2,4]`  
**Explanation:**  
The linked list is represented as: `[1,2,3,4]`.  
For `n = 4`, the middle node (index `2`, value `3`) is removed. The resulting list is `[1,2,4]`.

### Example 3
**Input:**  
`head = [2,1]`  
**Output:**  
`[2]`  
**Explanation:**  
The linked list is represented as: `[2,1]`.  
For `n = 2`, the middle node (index `1`, value `1`) is removed. The resulting list is `[2]`.

## Constraints
- The number of nodes in the list is in the range `[1, 10^5]`.
- `1 <= Node.val <= 10^5`


# Notes

1. The way to iterate is to use:
2. `while fast and fast.next`

# Submitted

```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def deleteMiddle(self, head: Optional[ListNode]) -> Optional[ListNode]:
        
        if head is None or head.next is None:
            return None

        fast = head
        slow = head

        dummy      = ListNode(0)
        dummy.next = head
        prev       = dummy

        while fast and fast.next:

            fast = fast.next.next
            slow = slow.next
            prev = prev.next


        prev.next = slow.next

        return dummy.next
```
