# 201. Bitwise AND of Numbers Range

# Description

Given two integers `left` and `right` that represent the range `[left, right]`, return the bitwise AND of all numbers in this range, inclusive.

# Examples

## Example 1
**Input:**  
`left = 5`, `right = 7`  
**Output:**  
`4`

## Example 2
**Input:**  
`left = 0`, `right = 0`  
**Output:**  
`0`

## Example 3
**Input:**  
`left = 1`, `right = 2147483647`  
**Output:**  
`0`

# Constraints

- `0 <= left <= right <= 2^31 - 1`



```python

class Solution:
    def rangeBitwiseAnd(self, m: int, n: int) -> int:
        
        def print_binary(description, input_binary_number):
            print(description, format(input_binary_number, '0{}_b'.format(32 + 9)))

        
        # print(format(n, '0{}_b'.format(width+3)))
        print_binary("m:", m)
        print_binary("n:", n)
        print("*" * 40)
        print("*" * 40)
        shift = 0
        # find the common 1-bits
        while m < n:
            m = m >> 1
            n = n >> 1
            shift += 1
            print_binary("m:", m)
            print_binary("n:", n)
            print("*" * 40)
            print("*" * 40)
        return m << shift
        
```

```python

class Solution:
    def rangeBitwiseAnd(self, m: int, n: int) -> int:
        
        def print_binary(description, input_binary_number):
            print(description, format(input_binary_number, '0{}_b'.format(32 + 9)))

        
        # print(format(n, '0{}_b'.format(width+3)))
        print_binary("m:", m)
        print_binary("n:", n)
        print("*" * 40)
        print("*" * 40)

        while m < n:
            # turn off rightmost 1-bit
            print_binary("n:", n)
            print_binary("n - 1:", n - 1)
            n = n & (n - 1)
            print_binary("n:", n)
            print("*" * 40)
            print("*" * 40)

            
        return n
        
```
