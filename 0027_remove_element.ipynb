# Problem 27: Remove Element

## Description
Given an integer array `nums` and an integer `val`, remove all occurrences of `val` in `nums` **in-place**. The order of the elements may be changed. Then return the number of elements in `nums` that are not equal to `val`.

To be accepted, your implementation must:
1. Modify the array `nums` such that the first `k` elements of `nums` contain the elements that are not equal to `val`. 
2. The remaining elements of `nums` are irrelevant and can be discarded.
3. Return the value `k`, representing the number of elements not equal to `val`.

### Custom Judge
The judge will test your solution with the following steps:
1. Initialize the input array `nums` and the value `val`.
2. Compare your output `k` with the expected number of elements.
3. Sort the first `k` elements of `nums` and verify they match the expected values.

If all assertions pass, your solution will be accepted.

## Examples

### Example 1
**Input:**  
`nums = [3,2,2,3]`  
`val = 3`  

**Output:**  
`2, nums = [2,2,_,_]`

**Explanation:**  
The function should return `k = 2`, with the first two elements of `nums` being `2`. The order of the first `k` elements does not matter. Values beyond `k` are ignored.

---

### Example 2
**Input:**  
`nums = [0,1,2,2,3,0,4,2]`  
`val = 2`  

**Output:**  
`5, nums = [0,1,4,0,3,_,_,_]`

**Explanation:**  
The function should return `k = 5`, with the first five elements of `nums` being `[0, 1, 4, 0, 3]` in any order. The values beyond `k` are irrelevant.

```python
def remove_element(nums):

    i = 0
    n = len(nums)

    while i < n:
    
        if nums[i] == val:
            nums[i] = nums[n - 1]
            n -= 1
        else:
            i += 1
    return n, nums
        
```
