```python
def get_digits_as_list(n):

    digits = []

    while n > 0:
        n, digit = divmod(n, 10)
        digits.append(str(digit))

    return list(reversed(digits))
```

```python
chars = ["a","a","b","b","c","c","c","c","c","c","c","c","c","c","c","c","d"] + ["."]
new_chars = chars[::]

i_char = 0
update_pos = 0

while i_char < ((len(chars)) - 1):

    repeat_count = 0
    i_rep = 0

    if chars[i_rep + i_char] == chars[i_rep + i_char + 1] and ((i_rep + i_char) < len(chars) - 1):

        # repetition case
        # print("Reptition Case")
        # print(f"{i_rep=}")
        # print(f"{i_char=}")
        # print(f"{chars[i_rep + i_char]=}")
        

        while (chars[i_rep + i_char] == chars[i_rep + i_char + 1]) and ((i_rep + i_char) < len(chars) - 1):

            i_rep += 1
            # print("Reptition Case")
            # print(f"{i_rep=}")
            # print(f"{chars[i_rep + i_char]=}")
            # print(f"{i_rep=}")
            # print(f"{i_char=}")
            
        new_chars[update_pos] = chars[i_char]
        if (i_rep + 1) > 9:
            digits = get_digits_as_list(i_rep + 1)
            for i, digit in enumerate(digits):
                new_chars[update_pos + i + 1] = digit
            update_pos += len(digits) + 1
        else:
            new_chars[update_pos + 1] = str(i_rep + 1)
            update_pos += 2
        i_char += 1 + i_rep
        # print("".join(new_chars))
    else:
        # no repetition case
        new_chars[update_pos] = chars[i_char]
        # print("".join(new_chars))
        i_char += 1
        update_pos += 1
```
