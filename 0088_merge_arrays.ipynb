# Deciding on the Iteration Method

1. We keep three pointers
    1. One for the writing position of nums1, one for end of nums2 and one for the reading position of nums1
2. When writing position of nums1 is < 0 then we stop.

```python
nums1 = [1, 2, 3, 0, 0, 0]
nums2 = [2, 5, 6]
```

```python
nums1 = [4, 5, 6, 0, 0, 0]
nums2 = [1, 2, 3]
```

```python
# nums1 = [0, 0, 0] # length m + n
# nums2 = [2, 5, 6] # length n
```

```python
# nums1 = [2, 0] # length m + n
# nums2 = [1] # length n
```

```python

n = len(nums2)
m = len(nums1) - n

p1 = m - 1
p2 = n - 1
p3 = m + n - 1
```

```python
print(f"{m=}, {n=}, {p1=}, {p2=}, {p3=}")
```

1. At every iteration, you read an element from the end of nums1 and nums2, and you write to a particular location in nums1

You read from p1 and p2 and you write to p3

1. If the element you read from p2 is greater than p1, then you write p2 element otherwise you write p1 element.
2. Whatever you write, you decrement the corresponding pointer

```python
special_case = 0

if n == 0:
    pass
else:
    if m == 0:
        for i in range(n):
            nums1[i] = nums2[i]
    else:
        while p3 >= 0:
            
            print("Inside first block")
            print(f"{m=}, {n=}, {p1=}, {p2=}, {p3=}")

            if (p1 < 0):
                if (p2 >= 0):
                    print("Inside Special Case 1 block")
                    print(f"{m=}, {n=}, {p1=}, {p2=}, {p3=}")
                    nums1[p3] = nums2[p2]
                    print(f"Wrote {nums1[p3]} at {p3} from {nums2[p2]}")
                    p2 = p2 - 1
                    p3 = p3 - 1
                    special_case = 1
                    print(f"After {m=}, {n=}, {p1=}, {p2=}, {p3=}")
                

            if (p2 < 0):
                if (p1 >= 0):
                    print("Inside Special Case 2 block")
                    nums1[p3] = nums1[p1]
                    p1 = p1 - 1
                    p3 = p3 - 1
                    special_case = 1
                
            if (special_case == 0) and (p1 >= 0) and (nums1[p1] >= nums2[p2]):
                
                print("Inside second block")
                print(f"{m=}, {n=}, {p1=}, {p2=}, {p3=}")
                
                nums1[p3] = nums1[p1]
                print(f"Wrote {nums1[p3]} at {p3} from {nums1[p1]}")
                
                p1 = p1 - 1
                p3 = p3 - 1
                
            elif (special_case == 0) and (p2 >= 0) and (nums1[p1] < nums2[p2]):
                
                print("Inside third block")
                nums1[p3] = nums2[p2]
                print(f"Wrote {nums1[p3]} at {p3} from {nums2[p2]}")
                print(f"{m=}, {n=}, {p1=}, {p2=}, {p3=}")
                p2 = p2 - 1
                p3 = p3 - 1
```
