```python
from collections import deque

# Define the Node class for our binary tree.
class Node:
    def __init__(self, data):
        self.data = data      # The value stored in the node.
        self.left = None      # Pointer to the left child.
        self.right = None     # Pointer to the right child.

    def __repr__(self):
        return str(self.data)
        # print(self.left)
        # print(self.right)
        
# Function to perform a level-order traversal and return a list of lists.
def level_order_traversal(root):
    # This will hold the final result.
    result = []
    
    # If the tree is empty, return an empty list.
    if root is None:
        return result

    # We use a queue to traverse the tree level by level.
    queue = deque([root])

    # Continue until there are no more nodes to process.
    while queue:
        print(f"Start of While loop: {queue=}")
        # Determine the number of nodes at the current level.
        level_size = len(queue)
        # This list will hold the nodes of the current level.
        current_level = []

        # Process all nodes for this level.
        for i in range(level_size):
            print(f"{i=}")
            # Remove the node at the front of the queue.
            node = queue.popleft()
            print(f"Start of for loop: {queue=}, {current_level=}")
            # Add its value to the current level list.
            current_level.append(node.data)

            # If the node has a left child, add it to the queue.
            if node.left:
                queue.append(node.left)
            print(f"After appending left node: {queue=}, {current_level=}")
            # If the node has a right child, add it to the queue.
            if node.right:
                queue.append(node.right)
            print(f"After appending right node: {queue=}, {current_level=}")
        print("=" * 80)
        print("=" * 80)
        print("=" * 80)

        # Append the current level list to the result.
        result.append(current_level)

    return result
```

```python
# Example usage and sample inputs:
# Example 1:
# Creating the following binary tree:
#
#          1
#        /   \
#       2     3
#      / \     \
#     4   5     6
#
root1 = Node(1)
root1.left = Node(2)
root1.right = Node(3)
root1.left.left = Node(4)
root1.left.right = Node(5)
root1.left.right.left = Node(15)
root1.left.right.right = Node(51)
root1.left.right.right.left = Node(151)
root1.right.right = Node(6)
root1.right.right.right = Node(7)

print("Example 1 - Tree with multiple levels:")
# Expected output: [[1], [2, 3], [4, 5, 6]]
print(level_order_traversal(root1))
print()
```

```python
# Example 2:
# A tree with only a single node.
#
#       10
#
root2 = Node(10)

print("Example 2 - Single Node Tree:")
# Expected output: [[10]]
print(level_order_traversal(root2))
print()
```

```python
# Example 3:
# An empty tree (no nodes at all).
root3 = None

print("Example 3 - Empty Tree:")
# Expected output: []
print(level_order_traversal(root3))
```

```python
class Solution:
    def lengthOfLIS(self, nums: List[int]) -> int:
        dp = [1] * len(nums)
        for i in range(1, len(nums)):
            for j in range(i):
                if nums[i] > nums[j]:
                    dp[i] = max(dp[i], dp[j] + 1)
                    print(f"{i=}, {j=}, {nums[i]=}, {nums[j]=}, {dp[i]=}")
                    for k in range(i + 1):
                        print(f"dp[{k}]: {dp[k]}")
                else:
                    print(f"Else: {i=}, {j=}, {nums[i]=}, {nums[j]=}, {dp[i]=}")
                    for k in range(i + 1):
                        print(f"dp[{k}]: {dp[k]}")

        return max(dp), dp
```
