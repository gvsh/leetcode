# Problem

# 122. Best Time to Buy and Sell Stock II

**Medium**


---

# Description

You are given an integer array `prices` where `prices[i]` is the price of a given stock on the `i`th day.

On each day, you may decide to buy and/or sell the stock. You can only hold at most one share of the stock at any time. However, you can buy it and then immediately sell it on the same day.

Find and return the maximum profit you can achieve.

---

# Examples

## Example 1:

**Input:**  
`prices = [7,1,5,3,6,4]`  

**Output:**  
`7`  

**Explanation:**  
- Buy on day 2 (price = 1) and sell on day 3 (price = 5), profit = `5-1 = 4`.
- Then buy on day 4 (price = 3) and sell on day 5 (price = 6), profit = `6-3 = 3`.  
- Total profit is `4 + 3 = 7`.

---

## Example 2:

**Input:**  
`prices = [1,2,3,4,5]`  

**Output:**  
`4`  

**Explanation:**  
- Buy on day 1 (price = 1) and sell on day 5 (price = 5), profit = `5-1 = 4`.  
- Total profit is `4`.

---

## Example 3:

**Input:**  
`prices = [7,6,4,3,1]`  

**Output:**  
`0`  

**Explanation:**  
- There is no way to make a positive profit, so we never buy the stock to achieve the maximum profit of `0`.

---

# Constraints:

- $ 1 <= prices.length <= 3 * 10^4$
- $ 0 <= prices[i] <= 10^4$

# Approach

1. We use the greedy approach, every time the price increases from the previous day, we buy and add it to our profit.
2. If we continue to do this for all days, we would get the maximum profit given the problem constraints.

# Algorithm

1. Initialization:
    2. `max_profit = 0` 
3. Iteration
    3. Compare price today to price yesterday, if it's positive, add the difference to profit.
4. Return: `max_profit`

# Example Walkthrough

1. Initialization:
    2. `prices = [7,1,5,3,6,4]`
    3. profit = 0
3. Iterations:
    4. Iteration 0:
        5. Index 0: Skip
        5. Index 1: index 1: 1 - 7 is negative, do nothing
        6. Index 2: index 3 - index 2: 5 - 1 = 4, add 4 to profit

```python
def best_time(prices):

    profit = 0

    for i in range(1, len(prices)):
        p_i = prices[i] - prices[i - 1]
        if p_i >= 0:
            profit += p_i

    return profit
```
